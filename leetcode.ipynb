````python
class Solution:
    def coinChange(self, coins: List[int], amount: int) -> int:
        # 先从子问题开始想，假设amount=0, 则需要0个coin, 所以initial state为dp[0] = 0, dp[i] = float("INF") for i > 0 ---> dp = [0, inf, inf, inf, ...]
        
        # 对于amount=11， coins = [1, 2, 5]时， 如果用1， dp[11] = dp[10] + 1, 如果用2， dp[11] = dp[9] + 1, 如果用5， dp[11] = dp[6] + 1, 
        # 所以 state transition formula is: dp[i] = min(dp[i-coin]+1 for coin in coins)
        

        dp = [float("inf")] * (amount + 1)
        dp[0] = 0

        # dp[amount] = min(dp[amount], dp[amount - coin] + 1 for coin in coins)

        for i in range(1, amount + 1):
            for coin in coins:
                if i -coin >= 0:
                    dp[i] = min(dp[i], dp[i-coin] + 1)

        ans = dp[amount]
        return ans if ans < float("inf") else -1

        
        #=========================================
        #从大到小找
        dp = {}

        def f(i, amount):
            if i < 0:
                return 0 if amount == 0 else float("INF")

            if (i, amount) in dp:
                return dp[(i, amount)]

            if coins[i] > amount: #大于当前金额，不能选, 如果是从小到大找，则不是这个判断
                return f(i-1, amount) # return也不一样

            select_this = f(i, amount - coins[i]) + 1
            no_select_this = f(i-1, amount)
            
            dp[(i, amount)] = min(select_this, no_select_this)
            return dp[(i, amount)]

        ans = f(len(coins) - 1, amount)
        return ans if ans < float("INF") else -1

        #========================================
        
        #从小到大找
        dp = {}
        def f(i, amount):
            if (i, amount) in dp:
                return dp[(i, amount)]
            if i == len(coins):
                if amount == 0:
                    return 0
                else:
                    return float("INF")

            if amount < 0: # amount < 0, 不能选，如果是从大到小找，则不是这个判断
                return float("INF") # return也不一样

            select_i = f(i, amount - coins[i]) + 1
            no_select = f(i+1, amount)

            dp[(i, amount)] = min(select_i, no_select)
            return dp[(i, amount)]

        ans = f(0, amount)
        return ans if ans <float("inf") else -1
````

````python
class Solution:
    def search(self, nums: List[int], target: int) -> int:
        # Divide array into two parts using mid
        # One part must be sorted, the other part then is rotated (Unless the original array is not rotated)
        # If the sorted part contains target so normal binary search

        # Else target must be in the other half (which may be rotated or sorted)
        #       We continue dividing, and the rotated part will again split into
        #       one sorted and one rotated part
        
        
        left = 0
        right = len(nums) - 1

        while left <= right:
            mid = (left + right) // 2

            if nums[mid] == target:
                return mid
            if nums[left] <= nums[mid]:

                if nums[left] <= target <= nums[mid]: # This does not use slice so no extra space
                    right = mid - 1
                else:
                    left = mid + 1

            else:
                if nums[mid] <= target <= nums[right]:
                    left = mid + 1
                else:
                    right = mid - 1
        return -1
    
        # ==========================================

        def bs(nums, target):
            left = 0
            right = len(nums) - 1

            while left <= right:
                mid = (right - left) // 2 + left
                if nums[mid] < target:
                    left = mid + 1
                elif nums[mid] > target:
                    right = mid - 1
                else:
                    return mid
            return -1

        left = 0
        right = len(nums) - 1

        while left <= right:
            mid = (right - left) // 2 + left
            if nums[left] <= nums[mid]: # Left part is sorted
                if nums[left] <= target <= nums[mid]: # Target is in left part
                    idx = bs(nums[left:mid+1], target) # This uses slice, extra space O(n)
                    return left + idx if idx != -1 else -1

                else: # Target is in right part, but right part is not sorted
                    left = mid + 1

            else: # Right part is sorted
                if nums[mid] <= target <= nums[right]: # target is in right part
                    idx = bs(nums[mid:right+1], target)
                    return mid + idx if idx != -1 else -1
                else:
                    right = mid - 1

        return -1
````

````python
'''
    假设nums1的长度是m, nums2的长度是n, 
    当m+n为奇数, 中位数是第(m+n)/2的数
    当m+n为偶数, 中位数是第(m+n)/2和第(m+n)/2+1的平均值
    
    所以当 m+n 为奇数时，中位数是合并后数组的第 ⌊(m+n)/2⌋ 个元素（从 0 开始计数
    当 m+n 为偶数时，中位数是合并后数组的第 ⌊(m+n)/2⌋ 和第 ⌊(m+n)/2⌋ + 1 个元素的平均值
    
    定义 i 为 nums1 左部分的元素数量, nums1[0...i-1] 为左部分, nums1[i...m-1] 为右部分
    定义 j 为 nums2 左部分的元素数量, nums2[0...j-1] 为左部分, nums2[j...n-1] 为右部分
    
    因此两个条件找到中位数:
    1.
    奇数长度时，左部分元素数量为 (m+n+1)/2, 右部分为 (m+n-1)/2
    偶数长度时，左右部分各有 (m+n)/2 个元素
    左边长度又可以用i+j表示, 则i+j = (m+n+1) // 2, 中位数为左边最大 (偶数也用这个, 计算的时候取左边最大右边最小的平均)
    则j = (m+n+1) // 2 - i
    而i可以用二分查找找到
    
    2.
    nums1[i-1] <= nums2[j](nums1 左部分最大值小于等于 nums2 右部分最小值)
    nums2[j-1] <= nums1[i](nums2 左部分最大值小于等于 nums1 右部分最小值)
    
    比如:
    奇数时
    nums1 = [1, 3, 5, 7, 9] m = 5
    nums2 = [2, 4, 6, 8] n = 4
    m+n = 9
    左部分元素数量：(m+n+1)/2 = 5, 右部分: (m+n-1)/2 = 4

    迭代i的值(二分查找), j = (m+n+1)//2 - i, 直到满足条件1和条件2
    则i=3, j=2, i+j = 3+2 = 5, (m-i)+(n-j) = (5-3)+(4-2) = 4
    nums1 左部分: [1, 3, 5]  ，右部分: [7, 9]
    nums2 左部分: [2, 4]     ，右部分: [6, 8]
    中位数: 左部分最大值 max(5, 4) = 5
    
    偶数时
    nums1 = [1, 3, 5, 7] m = 4
    nums2 = [2, 4, 6, 8] n = 4
    m+n = 8
    左部分元素数量：(m+n)/2 = 4, 右部分: (m+n)/2 = 4

    则i=2, j=2, i+j = 2+2 = 4, (m-i)+(n-j) = (5-2)+(4-2) = 4
    nums1 左部分: [1, 3]     ，右部分: [5, 7]
    nums2 左部分: [2, 4]     ，右部分: [6, 8]
    中位数: (左部分最大值 + 右部分最小值) / 2 = (max(3, 4) + min(5, 6)) / 2 = (4 + 5) / 2 = 4.5
    
    边界条件:
    i == n, j == m, 值为inf
    i == 0, j == 0, 值为-inf

    
'''
class Solution:
    def findMedianSortedArrays(self, nums1: List[int], nums2: List[int]) -> float:
        #确保在最短上找，减少时间复杂度，并且如果nums1比nums2长，i的值可能超过nums1的长度
        if len(nums1) > len(nums2):
            return self.findMedianSortedArrays(nums2, nums1)

        m, n = len(nums1), len(nums2)

        left = 0
        right = m

        while left <= right:
            i = (left+right) // 2
            j = (m+n+1)//2 - i

            nums1_left = -float("INF") if i == 0 else nums1[i-1]
            nums2_left = -float("INF") if j == 0 else nums2[j-1]

            nums1_right = float("INF") if i == m else nums1[i]
            nums2_right = float("INF") if j == n else nums2[j]

            if nums1_left <= nums2_right and nums2_left <= nums1_right:
                if not (m+n) % 2: #even
                    return float((max(nums1_left, nums2_left) + min(nums1_right, nums2_right)) / 2)

                else: #odd
                    return max(nums1_left, nums2_left)

            elif nums1_left > nums2_right:
                right = i - 1
            else:
                left = i + 1
````

````python
class Solution:
    def minCostClimbingStairs(self, cost: List[int]) -> int:
        
        f0 = 0
        f1 = 0

        for i in range(2, len(cost)+1):
            take1step = f1 + cost[i-1]
            take2step = f0 + cost[i-2]
            ans = min(take1step, take2step)
            f0 = f1
            f1 = ans

        return f1
    
#==========================================
        # Why len(cost)+1? Because we need to reach the step after the last step
        f = [0] * (len(cost)+1)

        for i in range(2, len(cost)+1):
            take1step = f[i-1] + cost[i-1]
            take2step = f[i-2] + cost[i-2]
            f[i] = min(take1step, take2step)

        return f[-1]

#==========================================
        dp = {}
        def f(i):
            if i in dp:
                return dp[i]

            # Why i <= 1 not i < 0? Because we can start from step 0 or step 1 directly, that means, we can reach step 0 or step 1 with 0 cost according to the problem description.
            if i <= 1:
                return 0

            select1 = f(i-1) + cost[i-1]
            select2 = f(i-2) + cost[i-2]

            dp[i] = min(select1, select2)
            return dp[i]
        return f(len(cost))
````

````python
class Solution:
    def lastStoneWeightII(self, stones: List[int]) -> int:
        # ------------------- @1 --------------------
        
        # This problem is very similar to 416 Partition Equal Subset Sum.
        #
        # Smashing stones is equivalent to assigning each stone either a positive
        # or negative sign, and we want the absolute final sum to be as small as possible.
        #
        # Another way to view it:
        # split stones into two groups. If one group's sum is x, the other group's
        # sum is total - x. The final remaining weight is:
        #
        #     abs((total - x) - x) = abs(total - 2 * x)
        #
        # To minimize this difference, we want x to be as close to total / 2 as possible.
        #
        # Therefore, this becomes a 0/1 knapsack problem:
        # find the maximum subset sum that is <= total // 2.
        # We focus on the smaller group and try to maximize its sum without exceeding half.

        total = sum(stones)
        s = total // 2

        dp = {}

        def f(i, c):
            # f(i, c) means:
            # using stones[0...i], given that the current chosen sum is c,
            # return the maximum sum we can finally reach without exceeding s.
            #
            # Different from the common 416 recursion:
            # - in 416, c usually means remaining capacity / target still needed
            # - here, c means current chosen sum, starting from 0, or you can say how much capacity left

            if (i, c) in dp:
                return dp[(i, c)]

            # No stones left, so the final chosen sum is c
            if i < 0:
                return c

            # Option 1: do not put stones[i] into the smaller group
            not_take = f(i - 1, c)

            # Option 2: put stones[i] into the smaller group,
            # only if the chosen sum does not exceed half
            if c + stones[i] <= s:
                take = f(i - 1, c + stones[i])
                dp[(i, c)] = max(take, not_take)
            else:
                dp[(i, c)] = not_take

            return dp[(i, c)]

        best = f(len(stones) - 1, 0)

        # best is the largest subset sum <= total // 2.
        # The other group has sum total - best.
        # Final difference = (total - best) - best = total - 2 * best.
        return total - 2 * best
    
    # ==========================================
    
        total = sum(stones)
        s = total // 2

        dp = [[0] * (s+1) for _ in range(len(stones)+1)]

        for c in range(s + 1):
            dp[0][c] = c

        for i in range(1, len(stones)+1):
            for c in range(s+1):
                if c + stones[i-1] <= s:
                    dp[i][c] = max(dp[i-1][c], dp[i-1][c+stones[i-1]])
                else:
                    dp[i][c] = dp[i-1][c]

        best = dp[len(stones)][0]
        return total - 2 * best
    
    # ===========================================
        # Because, our c means current capacity, so from 2d into 1d, we do not need to reverse the loop
        total = sum(stones)
        s = total // 2

        dp = [0] * (s+1)

        for c in range(s + 1):
            dp[c] = c

        for i in range(1, len(stones)+1):
            for c in range(s+1):
                if c + stones[i-1] <= s:
                    dp[c] = max(dp[c], dp[c+stones[i-1]])

        best = dp[0]
        return total - 2 * best
    
    # ---------------- @2 ---------------------------
    # Last one we define c as current capacity, now we define it as total capacity
    
        total = sum(stones)
        s = total // 2

        dp = {}

        def f(i, c):
            
            if i < 0:
                return 0

            if (i, c) in dp:
                return dp[(i, c)]

            if c >= stones[i]:
                no_take = f(i-1, c)
                take = f(i-1, c - stones[i]) + stones[i] # we add stones[i], so capacity reduces by stones[i]

                best = max(no_take, take)
            else:
                best = f(i-1, c)

            dp[(i, c)] = best
            return best

        return total - 2 * f(len(stones)-1, s)
    
    # ============================================
    
        total = sum(stones)
        s = total // 2

        dp = [[0] * (s+1) for _ in range(len(stones)+1)]

        for i in range(1, len(stones)+1):
            for c in range(1, s+1):
                if c >= stones[i-1]:
                    dp[i][c] = max(dp[i-1][c], dp[i-1][c-stones[i-1]] + stones[i-1])

                else:
                    dp[i][c] = dp[i-1][c]

        return total - 2 * dp[-1][-1]
    
    # ============================================
        total = sum(stones)
        s = total // 2

        dp = [0] * (s+1) 

        for i in range(1, len(stones)+1):
            for c in range(s, -1, -1):
                if c >= stones[i-1]:
                    dp[c] = max(dp[c], dp[c-stones[i-1]] + stones[i-1])

                else:
                    dp[c] = dp[c]

        return total - 2 * dp[-1]
````

````python
class DListNode:
    def __init__(self, key, val):
        self.key = key
        self.val  = val
        self.next = None
        self.prev = None


class LRUCache:
    def __init__(self, capacity: int):
        self.capacity = capacity
        self.dic = {}
        self.head = DListNode(-1, -1)
        self.tail = DListNode(-1, -1)
        self.head.next = self.tail
        self.tail.prev = self.head
        

    def get(self, key: int) -> int:
        if key not in self.dic:
            return -1

        node = self.dic[key]
        self.remove(node)
        self.add(node)
        return node.val

    def put(self, key: int, value: int) -> None:
        if key in self.dic:
            old_node = self.dic[key]
            self.remove(old_node)

        node = DListNode(key, value)
        self.dic[key] = node
        self.add(node)

        if len(self.dic) > self.capacity:
            node_to_delete = self.head.next
            self.remove(node_to_delete)
            del self.dic[node_to_delete.key]

    def add(self, node):
        previous_end = self.tail.prev
        previous_end.next = node
        node.prev = previous_end
        node.next = self.tail
        self.tail.prev = node
        
    def remove(self, node):
        node.prev.next = node.next
        node.next.prev = node.prev



# Your LRUCache object will be instantiated and called as such:
# obj = LRUCache(capacity)
# param_1 = obj.get(key)
# obj.put(key,value)
````

````python
class Solution:
    def longestPalindrome(self, s: str) -> str:
        # Dynamic programming solution, time complexity O(n^2), space complexity O(n^2)
        
        # dp[i][j] means s[i:j+1] is palindromic or not
        start = 0
        end = 0
        max_len = 0

        dp = [[False] * len(s) for _ in range(len(s))]
        
        for i in range(len(s)):
            dp[i][i] = True # "a" or "b" those single letter is also palindromic
                
        # dp[i][j] = dp[i+1][j-1] and s[i] == s[j]

        for i in range(len(s)-2, -1, -1):
            for j in range(i+1, len(s)):
                if s[i] == s[j]:
                    if j - i <= 2: # When s[i:j+1] is "a", "aa", "aba", then directly True
                        dp[i][j] = True
                    else:
                        dp[i][j] = dp[i+1][j-1]

                else:
                    dp[i][j] = False

                if dp[i][j] and j - i + 1 > max_len:
                    start = i
                    end = j
                    max_len = end-start + 1

        return s[start:end+1]
    
    # ===========================================
        ans_left = 0
        ans_right = 0

        # Odd situation
        for i in range(len(s)):
            l = i
            r = i

            while l >= 0 and r < len(s) and s[l] == s[r]:
                # When this stop, l and r is beyond the palindromic index, so actual index for palindromic is l+1 and r-1
                l-=1
                r+=1

            # No more palindromic, end and compare length of subtring
            if (r-1)-(l+1)+1 > ans_right-ans_left+1:
                ans_right = r-1
                ans_left = l+1

        # Even situation
        for i in range(len(s)):
            l = i
            r = i+1

            while l >= 0 and r < len(s) and s[l] == s[r]:
                # When this stop, l and r is beyond the palindromic index, so actual index for palindromic is l+1 and r-1
                l-=1
                r+=1

            # No more palindromic, end and compare length of subtring
            if (r-1)-(l+1)+1 > ans_right-ans_left+1:
                ans_right = r-1
                ans_left = l+1

        return s[ans_left:ans_right+1]
````

````python
class Solution:
    def lengthOfLIS(self, nums: List[int]) -> int:
        # In the greedy method, the while loop take O(n) time, so total time complexity is O(n^2)
        # We can use binary search to optimize the while loop to O(log n), so total time complexity is O(n log n)
        sub = [nums[0]]
        
        for num in nums[1:]:
            if num > sub[-1]:
                sub.append(num)
            else:
                # Find the first element in sub that is greater than or equal to num
                l = 0
                r = len(sub) - 1
                while l <= r:
                    mid = (l + r) // 2
                    if sub[mid] < num:
                        l = mid + 1
                    else:
                        r = mid - 1
                sub[l] = num

        return len(sub)
        
        
        # Greedy, O(n^2)
        sub = [nums[0]]
        
        for num in nums[1:]:
            if num > sub[-1]:
                sub.append(num)
            else:
                # Find the first element in sub that is greater than or equal to num
                i = 0
                while sub[i] < num:
                    i += 1
                sub[i] = num

        return len(sub)
        
        
        #======================================
        
        # Very initiative so no need recursion version
        # Time complexity is O(n^2)
        dp = [1] * len(nums)

        for i in range(len(nums)):
            for j in range(i):
                if nums[j] < nums[i]:
                    dp[i] = max(dp[i], dp[j] + 1)

        return max(dp)
    
````

````python
class Solution:
    def minimumTotal(self, triangle: List[List[int]]) -> int:
        dp = triangle[-1]
        
        for i in range(len(triangle)-2, -1, -1):
            for j in range(len(triangle[i])):
                dp[j] = min(dp[j], dp[j+1]) + triangle[i][j]

        return dp[0]
    
        # Bottom up, O(n^2), O(n^2) space
        
        dp = triangle.copy()
        
        for i in range(len(triangle)-2, -1, -1):
            for j in range(len(triangle[i])):
                dp[i][j] += min(dp[i+1][j], dp[i+1][j+1])

        return dp[0][0]
        
        #===========================
        
        # Top down, O(n^2), O(n^2) space

        dp = triangle.copy()
        
        for i in range(1, len(triangle)):
            for j in range(len(triangle[i])):
                if j == 0:
                    dp[i][j] += triangle[i-1][0]
                elif j < len(triangle[i-1]):
                    dp[i][j] += min(triangle[i-1][j], triangle[i-1][j-1])
                else:
                    dp[i][j] += triangle[i-1][j-1]

        return min(dp[-1])
````

````python
class Solution:
    def removeDuplicates(self, nums: List[int]) -> int:
        slow = 2
        fast = 2

        while fast < len(nums):
            if nums[fast] == nums[slow-2]:
                fast += 1
            else:
                nums[slow] = nums[fast]
                slow += 1
                fast += 1

        return slow
````

````python
class Solution:
    def rotate(self, nums: List[int], k: int) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        k %= len(nums) # Avoid case like [-1], k = 2

        # Do not use [::-1], because it will create a new list
        # Do not use nums[k:] + nums[:k], because it will create a new list
        # Do not use reversed(), because it will create a new iterator
        def reverse(left, right):
            while left < right:
                nums[left], nums[right] = nums[right], nums[left]
                left += 1
                right -= 1

        reverse(0, len(nums) - 1)
        reverse(0, k-1)
        reverse(k, len(nums)-1)
````

````python
class Solution:
    def jump(self, nums: List[int]) -> int:
        # We are gonna maintain two nxt, the current max reachable index, and the next max reachable index (Global Max)
        mx_nxt = 0 # This is the global max reachable (This global means from index 0 to i, what is the max reachable index, not 0 to end)
        cur_nxt = 0 # This is the current max reachable (This current means from the last jump, what is the max reachable index)
        ans = 0

        for i in range(len(nums)-1):
            mx_nxt = max(mx_nxt, i+nums[i])
            if i == cur_nxt: # If i equals to where we already reach, that means we reach the boundary, and we cannot move forward anymore, so we need to jump, but jump from where and to where?
                cur_nxt = mx_nxt # This line means: I cannot move forward anymore, so I GO BACK to where I can jump to the global max reachable, and I jump to it, so ans += 1
                ans += 1
        return ans
    
    # Here is some possible confusion:
    # 1. Why for i in range(len(nums)-1): not for i in range(len(nums)):
    # Because when we are at the last index, we do not need to jump anymore, so we do not need to consider the last index.
    
    # 2. Why greedy can find the optimal solution?
    # Because we are not greedy about single index, but all indices from last jump to current index, we are greedy about the global max reachable from all these indices, so we can always find the optimal solution.
````

````python
class Solution:
    def searchMatrix(self, matrix: List[List[int]], target: int) -> bool:

        def bs(nums: List[int], target: int) -> int:

            left = 0
            right = len(nums) - 1

            while left <= right:
                mid = left + (right - left) // 2
                if nums[mid] > target:
                    right = mid - 1
                elif nums[mid] < target:
                    left = mid + 1
                else:
                    return True
            return False

        left = 0
        right = len(matrix) - 1

        while left <= right:
            mid = (right-left) // 2 + left

            if matrix[mid][0] <= target and matrix[mid][-1] >= target:
                return bs(matrix[mid], target)
            elif matrix[mid][0] > target:
                right = mid - 1
            else:
                left = mid + 1
        return False
````

````python
class Solution:
    def isValidSudoku(self, board: List[List[str]]) -> bool:
        rows = [set() for _ in range(9)] 
        # Do not use [set() * 9] that means the element in set multiply by 9
        cols = [set() for _ in range(9)]
        subbox = [[set() for _ in range(3)]for _ in range(3)] 

        for r in range(len(board)):
            for c in range(len(board[0])):
                val = board[r][c]
                if val == '.':
                    continue
                
                # Check rows
                if val in rows[r]:
                    return False
                rows[r].add(val)
                
                # Check cols
                if val in cols[c]:
                    return False
                cols[c].add(val)
                
                # Check box
                if val in subbox[r//3][c//3]:
                    return False
                subbox[r//3][c//3].add(val)

        return True
````

````python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, x):
#         self.val = x
#         self.next = None

class Solution:
    def detectCycle(self, head: Optional[ListNode]) -> Optional[ListNode]:
        
        # head ---- a steps ---- entrance ---- b steps ---- meeting point
        #                  ↑                         |
        #                  |------------------------|
        
        # C is the whole cycle length
        # when slow and fast meet, slow moves a + b, fast moves 2(a+b) as its speed is as twice as slow's
        # Fast traveled one or more full cycles more than slow, so:
        # 2(a+b) - (a+b) = kc
        # a+b = kc
        # a = kc-b

        # This means:
        # The distance from head to entrance is the same as the distance from the meeting point back around the cycle to the entrance.
        # But we do not know kc, is 1c, 2c or 100c.

        # So we reset slow and keep fast at meeting point
        # Move slow and fast 1 step each time

        # After a steps, slow is at entrance, and fast is also because
        # a = (k - 1)C + (C - b)  ---> a steps = some full cycles + the final distance from meeting point to entrance
        # No matter how many cycle, fast always back to entrance with slow.
        
        # Check whatever is cycle or not
        slow = fast = head

        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next
            if slow == fast:
                # Find entrance
                slow = head
                while slow != fast:
                    slow = slow.next
                    fast = fast.next

                return slow

        return None


        # ====================
        slow = fast = head

        # Step 1: Check whether a cycle exists
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next

            if slow == fast:
                break
        else:
            return None

        # Step 2: Find the cycle entrance
        slow = head

        while slow != fast:
            slow = slow.next
            fast = fast.next

        return slow
````

````python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def reorderList(self, head: Optional[ListNode]) -> None:
        """
        Do not return anything, modify head in-place instead.
        """
        def find_mid(head):
            slow = fast = head

            while fast and fast.next:
                fast = fast.next.next
                slow =slow.next

            return slow


        def reverse_mid(head):
            pre = None
            cur = head

            while cur:
                nxt = cur.next
                cur.next = pre
                pre = cur
                cur = nxt

            return pre

        mid = find_mid(head)

        head1 = head
        head2 = reverse_mid(mid)

        while head2 and head2.next:
            nxt1 = head1.next
            nxt2 = head2.next

            head1.next = head2
            head2.next = nxt1

            head1 = nxt1
            head2 = nxt2

        return head
````

````python
class Solution:
    def canCompleteCircuit(self, gas: List[int], cost: List[int]) -> int:
        # There are several things needed to be considered:
        # 1. If the total gas is less than total cost, we cannot complete the circuit
        # 2. When we reach the bottom, we cannot go down more but go up, so once we find the bottom index, we start from the next index, and we cannot go lower.
        # 3. If total gas >= total cost, there must be a solution, and it is unique. 

        
        # =====================================
        
        # This is O(n^2) which is TLE
        n = len(gas)
        left = 0
        right = left + n - 1

        gas = gas + gas
        cost = cost + cost
        

        while left < n:
            use = 0
            ok = True
            for i in range(left, right + 1):
                use += gas[i] - cost[i]
                if use < 0:
                    ok = False
                    break
            if ok:
                return left
            left += 1
            right += 1

        return -1
````

  - **Sliding Window** is not working when there is negative number within the list, use **Prefix Sum** or **DP**

  - **Sliding Window** always follow this order:
    expand → shrink

    that means, you calculate value of first expanding, like s += nums[right], p *= nums[right]

    then you shrink, while p > k

    then count, ans += right-left+1 (count can before shrink, but shrink must after expand)

  ```
    def dfs():
        pass

  ```
  - When implementing **BFS**, we use deque like

  ```
    from collections import deque
  ```
    WHY? deque.popleft() is O(1), but list.pop(0) is O(n)

## Problem 3

Permutation testing is a modern, generic approach to testing statistical hypotheses. While traditional hypothesis testing is based on case-specific formulas and assumptions which don't always hold in practice, permutation testing provides a single, unified, non-parametric framework to perform all kinds of tests.

The idea is to randomly permute the data multiple times and calculate the quantity of interest each time. This way, we obtain the said quantity's distribution that captures variation that is produced by random chance. Then, we compare the quantity of interest we have observed in our data to the distribution of its random variation and calculate the p-value.

The goal of this task is to write a function that applies permutation testing to perform a chi-square test in order to compare the click-through rates of three website designs.

---

- the first row contains the number of clicks for each website design
- the second row specifies the number of user visits without a click

For example, in the following contingency table we have **10K user visits for each of the three website designs**, and for the first one **115 visits ended up with a click whilst 9885 users left without clicking**.

```

0   115   98  123

```

### 1. Compute expected values

Calculate the expected number of clicks and no-clicks assuming the three website designs generate the same number of clicks.

- expected clicks
- expected no-clicks

These are the averages of the corresponding rows in `cont_table`.

### 2. Compute chi-square statistic


The sum is taken over all **six values** from `cont_table`.

If the observed value comes from the **first row**, use the expected clicks.  
If the observed value comes from the **second row**, use the expected no-clicks.

````python
import numpy as np
import pandas as pd
# from random_clicks_generator import RandomClicksGenerator

def chi2_perm_test(cont_table: pd.DataFrame):
    # Step 1
    clicks = cont_table.iloc[0].to_numpy(dtype=int) # we can remove to numpy, then we get a series with index, to_numpy return a numpy array, which could be easier later. numpy array is faster, and support vector calculation
    non_clicks = cont_table.iloc[1].to_numpy(dtype=int)
    
    expected_clicks = clicks.mean()
    expected_non_clicks = non_clicks.mean()
    
    # Step 2
    chi2_clicks = 0
    for click in clicks:
        chi2_clicks += (click - expected_clicks) ** 2 / expected_clicks # You can also use ((clicks - expected_clicks) ** 2 / expected_clicks).sum(). Remember numpy array? This is vector calculation
        
    chi2_non_clicks = 0
    for non_click in non_clicks:
        chi2_non_clicks += (non_click - expected_non_clicks) ** 2 / expected_non_clicks
        
    chi2_obs = chi2_non_clicks + chi2_clicks
    
    # Step 3
    total_clicks = clicks.sum()
    total_non_clicks = non_clicks.sum()
    
    gen = RandomClicksGenerator(total_clicks, total_non_clicks)
    
    # Step 4
    rand_chi2_obs_list = []
    group_size = len(clicks) + len(non_clicks)
    for _ in range(1000):
        permuted_list = gen.shuffled_clicks()
        a = permuted_list[:group_size]
        b = permuted_list[group_size:group_size*2]
        c = permuted_list[group_size*2:]
        
        rand_clicks = np.array([a.sum(), b.sum(), c.sum()])
        rand_non_clicks = np.array([group_size - a.sum(), group_size - b.sum(), group_size - c.sum()])
        
        expected_rand_clicks = rand_clicks.mean()
        expected_rand_non_clicks = rand_non_clicks.mean()
        
        rand_chi2_obs = ((rand_clicks - expected_rand_clicks) ** 2 / expected_rand_clicks).sum() + ((rand_non_clicks - expected_rand_non_clicks) ** 2 / expected_rand_non_clicks).sum()
        
        rand_chi2_obs_list.append(rand_chi2_obs)
        
    # Step 5
    p_value = (np.array(rand_chi2_obs_list) >= np.array(chi2_obs)).mean().round(3)
    
    return {
        "expected": [expected_clicks, expected_non_clicks],
        "chi2_observed": chi2_obs,
        "chi2_random_mean": np.array(rand_chi2_obs_list).mean(),
        "p_value": p_value,
        "are_clicks_different": p_value < 0.05
    }
    
print(chi2_perm_test(cont_table))
````

## Scoring

Focus on **maximizing the number of successful allocations**.

Each test contains:

- a sequence of memory allocations and frees

## Assumptions

- `N` is a power of 2 in the range `[8..1024]`
- at most **1000 operations** (`alloc` / `free`)
- `free` will always be called on a currently allocated address

````python
class Allocator:

    def __init__(self, N: int):

        self.N = N

        # free lists for each block size
        self.bins = {}

        s = 1
        while s <= N:
            self.bins[s] = set()
            s = s * 2

        # initially the whole memory is free
        self.bins[N].add(0)
        # It looks like this
        # bins = {
            # 1: set(),
            # 2: set(),
            # 4: set(),
            # 8: {0}
        # }
        #Which means size of 1 to 8 free memory address
        
        # record allocated blocks
        self.allocd = {}

    def _ceil_pow2(self, x):
        # If we allocate 5 bytes, we actually allocate to 8 block size memory 
        s = 1
        while s < x:
            s = s * 2

        return s

    def alloc(self, size: int) -> int:

        need = self._ceil_pow2(size)

        s = need
        while s <= self.N and len(self.bins[s]) == 0:
            s = s * 2 # Go find the available block, if s is 4 but no 4 blocks, go to find 8, 16, 32

        if s > self.N:
            return -1

        addr = self.bins[s].pop() # Take out this addr, like 8:{0}, so addr is 8

        # split larger blocks
        while s > need: # Much larger than we need
            s = s // 2
            self.bins[s].add(addr + s)

        self.allocd[addr] = need

        return addr

    def free(self, address: int):

        s = self.allocd.pop(address)
        a = address

        while True:

            # find buddy without XOR
            if (a // s) % 2 == 0:
                buddy = a + s
            else:
                buddy = a - s

            if buddy not in self.bins[s]:
                break

            # merge with buddy
            self.bins[s].remove(buddy)

            a = min(a, buddy)
            s = s * 2

            if s > self.N:
                break

        self.bins[s].add(a)
````

````python
def solution1(N, M):
    max_side_length = int((4*N + M) ** 0.5)
    
    def ok(k):
        big = min(N, (k*k) // 4)
        small = k*k - big*4
        if small <= M:
            return True
        return False
    
    for k in range(max_side_length, -1, -1):
        if ok(k):
            return k
        
    return 0

def solution(M, N):

    def can(k):
        big = min(N, (k // 2) * (k // 2))
        small = k*k - big*4
        return small <= M

    left = 0
    right = int((M + 4*N)**0.5) + 1
    ans = 0

    while left <= right:
        mid = (left + right) // 2

        if can(mid):
            ans = mid
            left = mid + 1
        else:
            right = mid - 1

    return ans

def test():
    tests = [
        # (N, M, expected)
        (0, 8, 2),   # example 1
        (3, 4, 4),   # example 2
        (18, 0, 8),  # example 3
        (3, 13, 5),  # example 4
        (1, 0, 2),   # one 2x2 brick
        (0, 1, 1),   # only one 1x1
        (0, 3, 1),   # three 1x1 → max square 1
        (2, 0, 2),   # two 2x2 bricks → 2x2 square
    ]

    for N, M, expected in tests:
        result = solution1(N, M)
        print(f"N={N}, M={M}, result={result}, expected={expected}")
        if result != expected:
            return False

    return True


print(test())
````

#### ***HSBC_3***

### Example 1

S = "bababb"  
→ three-letter substrings: **bab**, **aba**, **bab**, **abb**  
→ distinct = {bab, aba, abb}  
**Output:** 3


S = "bbbaaab"  
→ substrings: **bbb**, **bba**, **baa**, **aaa**, **aab**  
→ distinct = 5  
**Output:** 5


S = "ab"  
→ no length-3 substring  
**Output:** 0

````python
def Solution(S):
    if len(S) < 3:
        return 0
    if len(S) == 3:
        return 1
    
    left = 0
    seen = set()
    
    while left < len(S)-2:
        right = left + 2
        substring = S[left:right+1]
        if substring not in seen:
            seen.add(substring)
            
        left += 1
            
    return len(seen)

print(Solution("bbbaaab"))
````

You are given two strings `S` and `T`, both of length `N`.  
Each string consists only of digits (`'0'–'9'`).

In a single move, you can choose an index `K` in the range `[0, N−2]` and **increase both digits `S[K]` and `S[K+1]` by 1**.  
If a digit is `9` and you increase it, it becomes `0` (mod 10).

Example:

Input: S = "13471", T = "59604"
Output: 9

S = "557", T = "445"

| Step | Operation | Result |
|:----:|:-----------|:--------|
| 0 | Initial | 557 |
| +1 | increase(0) | 667 |
| +2 | increase(0) | 778 |
| +3 | increase(0) | 889 |
| +4 | increase(0) | 990 |
| +5 | increase(0) | 001 |
| +6 | increase(0) | 112 |
| +7 | increase(0) | 223 |
| +8 | increase(0) | 334 |
| +9 | increase(0) | 445 |

Tips: 

- Every move must **act on index `i` and `i+1` together**.  
- The operation is **cyclic modulo 10**.  
- You need to find the **minimum number of moves** to transform `S` into `T`.

````python
def solution(S, T):
    S = [int(x) for x in S]
    T = [int(x) for x in T]
    n = len(S)
    ans = 0

    for i in range(n - 1):
        diff = (T[i] - S[i]) % 10
        S[i] = (S[i] + diff) % 10
        S[i + 1] = (S[i + 1] + diff) % 10
        ans += diff

    if S[-1] != T[-1]:
        return -1

    return ans
````

#### ***HSBC_7***

You are given a string `S`, which consists entirely of decimal digits (`0–9`).  
Using digits from `S`, create a **palindromic number** with the **largest possible decimal value**.  
You should use at least **one** digit and you **can reorder** the digits.

A palindromic number remains the same when its digits are reversed; for instance, `"7"`, `"44"` or `"83238"`.  
Any palindromic number you create **must not have leading zeros**, such as `"0990"` or `"010"`.

For example, decimal palindromic numbers that can be created from `"8199"` are:  
`"1"`, `"8"`, `"9"`, `"9"`, `"919"` and `"989"`.  
Among them, `"989"` has the largest value.

````python
def solution(S):
    cnt = {}
    for ch in S:
        cnt[ch] = cnt.get(ch, 0) + 1

    left_parts = []
    for i in range(9, -1, -1):
        d = str(i)
        pairs = cnt.get(d, 0) // 2
        if pairs > 0:
            left_parts.append(d * pairs)
            cnt[d] = cnt.get(d, 0) - pairs * 2

    centre = ''
    for i in range(9, -1, -1):
        d = str(i)
        if cnt.get(d, 0) % 2 == 1:
            centre = d
            break

    left_str = ''.join(left_parts)
    ans = left_str + centre + left_str[::-1]

    j = 0
    while j < len(ans) and ans[j] == '0':
        j += 1
        
    if j == len(ans):
        
        return '0'
    return ans[j:]
````
