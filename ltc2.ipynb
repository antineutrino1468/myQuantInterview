## LC 213 | House Robber II

- Topic: Dynamic Programming
- Main method: `rob`
- Link: [House Robber II](https://leetcode.com/problems/house-robber-ii/)


````python
class Solution:
    def rob(self, nums: List[int]) -> int:
        # if we steal nums[0], then we start with nums[0], and you never go to nums[-1]
        # if we do not steal nums[0], then we start with nums[1]

        def f(nums):
            if len(nums) <= 2:
                return max(nums)

            dp0 = nums[0]
            dp1 = max(nums[0], nums[1])

            for i in range(2, len(nums)):
                ans = max(dp1, dp0+nums[i])
                dp0 = dp1
                dp1 = ans

            return dp1
        if len(nums) == 1:
            return nums[0]

        return max(f(nums[:-1]), f(nums[1:]))
````

## LC 1223 | Dice Roll Simulation

- Topic: Dynamic Programming
- Main method: `dieSimulator`
- Link: [Dice Roll Simulation](https://leetcode.com/problems/dice-roll-simulation/)


````python
class Solution:
    def dieSimulator(self, n: int, rollMax: List[int]) -> int:
        @cache
        def f(i, last, left):
            # i is the number of rolls remaining; last is the previous die face; left is how many more times that face can be used consecutively.
            if i == 0:
                return 1

            res = 0

            for index in range(len(rollMax)):
                if index != last:
                    res += f(i-1, index, rollMax[index] - 1)
                elif left != 0:
                    res += f(i-1, index, left - 1)

            return res
        ans = 0
        for k in range(len(rollMax)):
            ans += f(n-1, k, rollMax[k]-1)

        return ans % 1000000007
````

````python
class Solution:
    def maxProfit(self, prices: List[int]) -> int:
        ans = 0
        cost = float("inf")

        for i in prices:
            if i < cost:
                cost = i

            if i-cost > ans:
                ans = i-cost
        return ans
````

## LC 1048 | Longest String Chain

- Topic: Dynamic Programming
- Main method: `longestStrChain`
- Link: [Longest String Chain](https://leetcode.com/problems/longest-string-chain/)


````python
class Solution:
    def longestStrChain(self, words: List[str]) -> int:
        dp = {}
        words.sort(key = len)

        for word in words:
            dp[word] = 1
            for i in range(len(word)):
                s = word[:i] + word[i+1:]
                # if s in words:
                dp[word] = max(dp[word], dp.get(s, 0) + 1)

        return max(dp.values())

        #===================================================

        dp = {}
        def f(s):
            if s in dp:
                return dp[s]

            res = 0
            for i in range(len(s)):
                sub = s[:i] + s[i+1:] # as the question said, we can only insert exactly one letter
                if sub in words:
                    res = max(res, f(sub))
            dp[s] = res + 1
            return res + 1

        ans = 0
        for i in words:
            ans = max(ans, f(i))
        return ans
````

## LC 70 | Climbing Stairs

- Topic: Dynamic Programming
- Main method: `climbStairs`
- Link: [Climbing Stairs](https://leetcode.com/problems/climbing-stairs/)


````python
class Solution:
    def climbStairs(self, n: int) -> int:
        if n <= 2:
            return n

        dp0 = 1
        dp1 = 2

        for i in range(2, n):
            ans = dp1 + dp0
            dp0 = dp1
            dp1 = ans

        return dp1
    # ===========================================
        if n <= 3:
            return n

        dp = [0] * n

        dp[0] = 1
        dp[1] = 2

        for i in range(2, n):
            dp[i] = dp[i-1] + dp[i-2]

        return dp[-1]



    # ===========================================

        dp = {}

        def f(i):
            if i in dp:
                return dp[i]

            if i <= 3:
                return i

            climb_1_step = f(i-1)
            climb_2_step = f(i-2)

            dp[i] = climb_1_step + climb_2_step
            return dp[i]

        return f(n)
````

## LC 322 | Coin Change

- Topic: Dynamic Programming
- Main method: `coinChange`
- Link: [Coin Change](https://leetcode.com/problems/coin-change/)


````python
class Solution:
    def coinChange(self, coins: List[int], amount: int) -> int:
        # Start from amount = 0: it needs 0 coins. Initialize dp[0] = 0 and dp[i] = infinity for i > 0.

        # Example: for amount = 11 and coins = [1, 2, 5], try dp[10] + 1, dp[9] + 1, and dp[6] + 1.
        # State transition: dp[i] = min(dp[i - coin] + 1 for coin in coins).


        dp = [float("inf")] * (amount + 1)
        dp[0] = 0

        # dp[amount] = min(dp[amount], dp[amount - coin] + 1 for coin in coins)

        for i in range(1, amount + 1):
            for coin in coins:
                if i -coin >= 0:
                    dp[i] = min(dp[i], dp[i-coin] + 1)

        ans = dp[amount]
        return ans if ans < float("inf") else -1


        #=========================================
        #
        dp = {}

        def f(i, amount):
            if i < 0:
                return 0 if amount == 0 else float("INF")

            if (i, amount) in dp:
                return dp[(i, amount)]

            if coins[i] > amount: #,
                return f(i-1, amount) # return

            select_this = f(i, amount - coins[i]) + 1
            no_select_this = f(i-1, amount)

            dp[(i, amount)] = min(select_this, no_select_this)
            return dp[(i, amount)]

        ans = f(len(coins) - 1, amount)
        return ans if ans < float("INF") else -1

        #========================================

        #
        dp = {}
        def f(i, amount):
            if (i, amount) in dp:
                return dp[(i, amount)]
            if i == len(coins):
                if amount == 0:
                    return 0
                else:
                    return float("INF")

            if amount < 0: # amount < 0,
                return float("INF") # return

            select_i = f(i, amount - coins[i]) + 1
            no_select = f(i+1, amount)

            dp[(i, amount)] = min(select_i, no_select)
            return dp[(i, amount)]

        ans = f(0, amount)
        return ans if ans <float("inf") else -1
````

## LC 509 | Fibonacci Number

- Topic: Dynamic Programming
- Main method: `fib`
- Link: [Fibonacci Number](https://leetcode.com/problems/fibonacci-number/)


````python
class Solution:

    def fib(self, n: int) -> int:
        if n == 0:
            return 0
        if n == 1:
            return 1

        # dp[i] = dp[i-1] + dp[i-2]
        dp1 = 0
        dp2 = 1

        for i in range(1, n):
            ans = dp1 + dp2
            dp1 = dp2
            dp2 = ans

        return ans
````

## LC 494 | Target Sum

- Topic: Dynamic Programming
- Main method: `findTargetSumWays`
- Link: [Target Sum](https://leetcode.com/problems/target-sum/)


````python
class Solution:
    def findTargetSumWays(self, nums: List[int], target: int) -> int:
        # P means positive sum, N means negative sum
        # P - N = sum

        # P + N = target
        # P = target - N
        # P = target - P + sum
        # 2P = target + sum
        # P = (target+sum) // 2

        # so we select n num from nums where n == P

        s = sum(nums)
        p = (target+s)
        bag = p//2

        if (s + target) % 2 != 0 or abs(target) > s:
            return 0

        # Transition
        # dp[i] = dp[i] + dp[i-num] for num in nums

        dp = [0] * (bag + 1)
        dp[0] = 1
        for num in nums:
            for i in range(bag, num - 1, -1):
                    dp[i] = dp[i] + dp[i-num]

        return dp[bag]
    #=========================================

        # p + n = target
        # p - n = s
        # n = p-s
        # p + p - s = target
        # 2p = target + s
        # p = (target + s)//2

        s = sum(nums)
        if (target+s) % 2 != 0:
            return 0

        if s < abs(target):
            return 0

        capacity = (target+s)//2

        # dp[i][capacity] = dp[i-1][capacity-nums[i]] + dp[i-1][capacity]

        dp = [[0] * (capacity+1) for _ in range(len(nums)+1)]
        dp[0][0] = 1

        for i in range(1, len(nums)+1):
            for c in range(capacity+1):
                if c >= nums[i-1]:
                    dp[i][c] = dp[i-1][c-nums[i-1]] + dp[i-1][c]
                else:
                    dp[i][c] = dp[i-1][c]

        return dp[-1][-1]
    #=========================================
        # Time complexity: O(n * m). Space complexity: O(n * m).
        # top-down DP
        dp = {}

        def f(i, s):

            if (i, s) in dp:
                return dp[(i, s)]

            if i < 0:
                if s == 0:
                    return 1
                else:
                    return 0

            plus = f(i-1, s + nums[i])
            minus = f(i-1, s - nums[i])

            dp[(i, s)] = plus + minus
            return dp[(i, s)]

        return f(len(nums)-1, target)

        #bottom-up DP
        dp = {}

        def f(i, s):

            if (i, s) in dp:
                return dp[(i, s)]

            if i >= len(nums):
                if target == s:
                    return 1
                else:
                    return 0

            plus = f(i+1, s+nums[i])
            minus = f(i+1, s - nums[i])

            dp[(i, s)] = plus + minus
            return dp[(i, s)]

        return f(0, 0)
````

## LC 279 | Perfect Squares

- Topic: Dynamic Programming
- Main method: `numSquares`
- Link: [Perfect Squares](https://leetcode.com/problems/perfect-squares/)


````python
class Solution:
    def numSquares(self, n: int) -> int:

        n_root = []
        for i in range(1, n+1): # if n == 1, then nothing in list if range(i, n)
            if i*i <= n:
                n_root.append(i*i)
            else:
                break

        # dp[i][c] = min(dp[i][c - nums[i-1]] + 1, dp[i-1][c])
        dp = [float("inf")] * (n + 1)

        dp[0] = 0

        for i in range(1, len(n_root) + 1):
            for c in range(1, n + 1):
                if c >= n_root[i-1]:
                    dp[c] = min(dp[c-n_root[i-1]] + 1, dp[c])
                else:
                    dp[c] = dp[c]

        return dp[-1]

        # ======================================

        n_root = []
        for i in range(1, n+1): # if n == 1, then nothing in list if range(i, n)
            if i*i <= n:
                n_root.append(i*i)
            else:
                break

        # dp[i][c] = min(dp[i][c - nums[i-1]] + 1, dp[i-1][c])
        dp = [[float("inf")] * (n + 1) for _ in range(len(n_root) + 1)]

        for i in range(len(n_root) + 1): # For all c == 0, we do not need to select anything, so initialise it into 0
            dp[i][0] = 0

        for i in range(1, len(n_root) + 1):
            for c in range(1, n + 1):
                if c >= n_root[i-1]:
                    dp[i][c] = min(dp[i][c-n_root[i-1]] + 1, dp[i-1][c])
                else:
                    dp[i][c] = dp[i-1][c]

        return dp[-1][-1]

    # ========================================

        nums = [i**2 for i in range(1, n+1)]

        dp = [float("inf")] * (n+1)
        dp[0] = 0

        for c in range(1, n+1):
            for x in nums:
                if c >= x:
                    dp[c] = min(dp[c], dp[c-x] + 1)

        return dp[-1]

        # ========================================

        nums = [i**2 for i in range(1, n)]

        @cache
        def f(c):
            if c == 0:
                return 0

            if c < 0:
                return float("inf")

            ans = float("inf")
            for x in nums:
                if c >= x:
                    ans = min(ans, f(c-x)+1)

            return ans
        return f(n)
````

````python
class Solution:
    def findMaxForm(self, strs: List[str], m: int, n: int) -> int:

        dp = [[0] * (n + 1) for _ in range(m + 1)]

        for i in range(1, len(strs)+1):
            for c1 in range(m, -1, -1):
                for c2 in range(n, -1, -1):
                    zero = strs[i-1].count("0")
                    one = strs[i-1].count("1")
                    if c1 >= zero and c2 >= one:
                        dp[c1][c2] = max(dp[c1-zero][c2-one]+1, dp[c1][c2])
                    else:
                        dp[c1][c2] = dp[c1][c2]

        return dp[-1][-1]

    # =========================================
        dp = [[[0] * (n + 1) for _ in range(m + 1)] for _ in range(len(strs) + 1)]

        for i in range(1, len(strs)+1):
            for c1 in range(m+1):
                for c2 in range(n+1):
                    zero = strs[i-1].count("0")
                    one = strs[i-1].count("1")
                    if c1 >= zero and c2 >= one:
                        dp[i][c1][c2] = max(dp[i-1][c1-zero][c2-one]+1, dp[i-1][c1][c2])
                    else:
                        dp[i][c1][c2] = dp[i-1][c1][c2]

        return dp[-1][-1][-1]
    # ==========================================
        @cache
        def f(i, c1, c2):
            if i < 0:
                return 0

            zero = strs[i].count("0")
            one = strs[i].count("1")

            no_select = f(i-1, c1, c2)
            if c1 >= zero and c2 >= one:
                select = f(i-1, c1-zero, c2-one) + 1
                return max(select, no_select)
            else:
                return no_select

        return f(len(strs)-1, m, n)
````

## LC 1035 | Uncrossed Lines

- Topic: Dynamic Programming
- Main method: `maxUncrossedLines`
- Link: [Uncrossed Lines](https://leetcode.com/problems/uncrossed-lines/)


````python
class Solution:
    def maxUncrossedLines(self, nums1: List[int], nums2: List[int]) -> int:
        # If we draw two lines nums1[i] to nums2[j], and nums1[x] to nums2[y]
        # If these two lines do not cross each other, i < j then x < y, if i > j then x > y
        # So actually this problem is as same as LCS

        # dp[i][j] = dp[i-1][j-1] + 1 or max(dp[i-1][j], dp[i][j-1])

        dp = [0] * (len(nums2)+1)

        for i in range(1, len(nums1)+1):
            prev = 0
            for j in range(1, len(nums2)+1):
                temp = dp[j]
                if nums1[i-1] == nums2[j-1]:
                    dp[j] = prev + 1

                else:
                    dp[j] = max(dp[j], dp[j-1])
                prev = temp
        return dp[-1]
````

## LC 72 | Edit Distance

- Topic: Dynamic Programming
- Main method: `minDistance`
- Link: [Edit Distance](https://leetcode.com/problems/edit-distance/)


````python
class Solution:
    def minDistance(self, word1: str, word2: str) -> int:
        '''
        word1 = "ab", word2 = "abc"

        i=1, j=1:
        dp = [0, 1, 2, 3]
        prev = 0
        dp -> [1, 1, 2, 3]
        temp = 1


            ""  a   b   c
        ""  0   1   2   3
        a   1   0   0   0
        b   2   0   0   0

        i=1, j=2:
        dp = [1, 1, 2, 3]
        prev = 1
        dp -> [1, 1, 2, 3]
        temp = 2


            ""  a   b   c
        ""  0   1   2   3
        a   1   0   1   0
        b   2   0   0   0

        i=1, j=3:
        dp = [1, 1, 2, 3]
        prev = 2
        dp -> [2, 1, 2, 3]
        temp = 3


            ""  a   b   c
        ""  0   1   2   3
        a   1   0   1   2
        b   2   0   0   0

        '''
        m, n = len(word1), len(word2)

        dp = [i for i in range(n+1)]

        # Delete a character from word1.
        for i in range(1, m + 1):
            # Use the top-left value dp[i - 1][j - 1] for replacement.
            prev = dp[0]
            dp[0] = i

            # Insert a character from word2.
            for j in range(1, n + 1):
                temp = dp[j] #  dp[j]  temp prev
                if word1[i-1] == word2[j-1]:
                    dp[j] = prev
                else:
                    # - dp[j] = dp[i - 1][j]: delete a character from word1.
                    # - dp[j - 1] = dp[i][j - 1]: insert a character from word2.
                    # - prev = dp[i - 1][j - 1]: replace a character.
                    dp[j] = min(dp[j], dp[j-1], prev) + 1
                prev = temp #  prev

        return dp[n]

        #==========================================
        '''
        word1 = "ab", word2 = "abc"
            ""  a   b   c
        ""  0   1   2   3
        a   1   0   0   0
        b   2   0   0   0

        i = 1 j = 1:

            ""  a   b   c
        ""  0   1   2   3
        a   1   0   0   0
        b   2   0   0   0

        i = 1, j = 2:
            ""  a   b   c
        ""  0   1   2   3
        a   1   0   1   0
        b   2   0   0   0

        i = 1, j = 3:
            ""  a   b   c
        ""  0   1   2   3
        a   1   0   1   2
        b   2   0   0   0
        .
        .
        .

            ""  a   b   c
        ""  0   1   2   3
        a   1   0   1   2
        b   2   1   0   1

        '''
        m, n = len(word1), len(word2)

        dp = [[0] * (n + 1) for _ in range(m + 1)]

        for j in range(len(word2) + 1):
            dp[0][j] = j
        for i in range(len(word1) + 1):
            dp[i][0] = i

        for i in range(1, m + 1):
            for j in range(1, n + 1):
                if word1[i-1] == word2[j-1]:
                    dp[i][j] = dp[i-1][j-1]
                else:
                    dp[i][j] = min(dp[i][j-1], dp[i-1][j], dp[i-1][j-1]) + 1

        return dp[m][n]

        #==========================

        # min(insert, delete, replace) = min(dp[i][j - 1], dp[i - 1][j], dp[i - 1][j - 1]) + 1.
        # Example: word1 = "ab", word2 = "abc". Insert "c" into word1; equivalently, consume/delete "c" from word2.
        # Example: word1 = "abc", word2 = "ab". Delete "c" from word1.
        # Example: word1 = "abc", word2 = "abd". Replace "c" with "d".

        dp = {}

        def f(i, j):
            if (i, j) in dp:
                return dp[(i, j)]

            if i < 0:
                return j+1 #  word1 = "abc"i = 2word2 = "" 'a', 'b', 'c' 3
            if j < 0:
                return i+1

            if word1[i] == word2[j]:
                return f(i-1, j-1)

            insert = f(i, j-1)
            delete = f(i-1, j)
            replace = f(i-1, j-1)

            dp[(i, j)] = min(insert, delete, replace) + 1
            return dp[(i, j)]

        return f(len(word1)-1, len(word2)-1)
````

## LC 53 | Maximum Subarray

- Main method: `maxSubArray`
- Link: [Maximum Subarray](https://leetcode.com/problems/maximum-subarray/)


````python
class Solution:
    def maxSubArray(self, nums: List[int]) -> int:
        # Prefix Sum
        min_prefix_sum = cur_prefix_sum = 0
        ans = -float("INF")

        for i in nums:
            cur_prefix_sum += i
            ans = max(ans, cur_prefix_sum - min_prefix_sum)
            min_prefix_sum = min(min_prefix_sum, cur_prefix_sum)

        return ans

        # DP
        current = nums[0]
        best = nums[0]

        for i in range(1, len(nums)):
            current = max(current, 0) + nums[i]
            best = max(current, best)

        return best

        #=========================================
        # Time complexity: O(n). Space complexity: O(n).
        dp = [0] * len(nums)
        dp[0] = nums[0]

        for i in range(1, len(nums)):
            dp[i] = max(dp[i-1], 0) + nums[i]

        return max(dp)
````

## LC 746 | Min Cost Climbing Stairs

- Topic: Dynamic Programming
- Main method: `minCostClimbingStairs`
- Link: [Min Cost Climbing Stairs](https://leetcode.com/problems/min-cost-climbing-stairs/)


````python
class Solution:
    def minCostClimbingStairs(self, cost: List[int]) -> int:

        f0 = 0
        f1 = 0

        for i in range(2, len(cost)+1):
            take1step = f1 + cost[i-1]
            take2step = f0 + cost[i-2]
            ans = min(take1step, take2step)
            f0 = f1
            f1 = ans

        return f1

#==========================================
        # Why len(cost)+1? Because we need to reach the step after the last step
        f = [0] * (len(cost)+1)

        for i in range(2, len(cost)+1):
            take1step = f[i-1] + cost[i-1]
            take2step = f[i-2] + cost[i-2]
            f[i] = min(take1step, take2step)

        return f[-1]

#==========================================
        dp = {}
        def f(i):
            if i in dp:
                return dp[i]

            # Why i <= 1 not i < 0? Because we can start from step 0 or step 1 directly, that means, we can reach step 0 or step 1 with 0 cost according to the problem description.
            if i <= 1:
                return 0

            select1 = f(i-1) + cost[i-1]
            select2 = f(i-2) + cost[i-2]

            dp[i] = min(select1, select2)
            return dp[i]
        return f(len(cost))
````

````python
class Solution:
    def uniquePaths(self, m: int, n: int) -> int:

        dp = [1] * n
        for _ in range(1, m):
            for j in range(1, n):
                dp[j] += dp[j - 1]
        return dp[-1]

#===========================================

        dp = [[0] * (n) for _ in range(m)]

        for i in range(m):
            dp[i][0] = 1
        for j in range(n):
            dp[0][j] = 1

        for i in range(1, m):
            for j in range(1, n):
                dp[i][j] = dp[i-1][j] + dp[i][j-1]

        return dp[m-1][n-1]

#===========================================
        dp = {}
        def f(i, j):
            if (i, j) in dp:
                return dp[(i, j)]

            if i == 0 and j == 0:
                return 1

            if i<0:
                return 0

            if j < 0:
                return 0

            down = f(i-1, j)
            right = f(i, j-1)

            dp[(i, j)] = down + right
            return dp[(i, j)]

        return f(m-1, n -1)
````

## LC 416 | Partition Equal Subset Sum

- Topic: Dynamic Programming
- Main method: `canPartition`
- Link: [Partition Equal Subset Sum](https://leetcode.com/problems/partition-equal-subset-sum/)


````python
class Solution:
    def canPartition(self, nums: List[int]) -> bool:

        s = sum(nums)
        if s % 2 != 0:
            return False

        target = s // 2

        dp = [False] * (target+1)
        dp[0] = True

        for i in range(1, len(nums)+1):
            for c in range(target, -1, -1):
                if c >= nums[i-1]:
                    dp[c] = dp[c - nums[i-1]] or dp[c]

        return dp[-1]

#==========================================

        s = sum(nums)
        if s % 2 != 0:
            return False

        target = s // 2

        # dp[i][capacity] = dp[i-1][capacity - nums[i-1]] + dp[i-1][capacity]

        dp = [[False] * (target+1) for _ in range(len(nums)+1)]
        dp[0][0] = True

        for i in range(1, len(nums)+1):
            for c in range(1, target+1):
                if c >= nums[i-1]:
                    dp[i][c] = dp[i-1][c - nums[i-1]] or dp[i-1][c]
                else:
                    dp[i][c] = dp[i-1][c]
        return dp[-1][-1]
#==========================================
        total = sum(nums)
        if total % 2 != 0:
            return False

        dp = {}

        def f(i, amount):
            if (i, amount) in dp:
                return dp[(i, amount)]

            if amount < 0:
                return False

            if i == 0:
                if amount == 0:
                    return True
                else:
                    return False


            dp[(i, amount)] = f(i-1, amount - nums[i]) or f(i-1, amount)
            return dp[(i, amount)]


        return f(len(nums) - 1, total/2)
````

- Main method: `lastStoneWeightII`
- Link: [Last Stone Weight II](https://leetcode.com/problems/last-stone-weight-ii/)


````python
class Solution:
    def lastStoneWeightII(self, stones: List[int]) -> int:
        # ------------------- @1 --------------------

        # This problem is very similar to 416 Partition Equal Subset Sum.
        #
        # Smashing stones is equivalent to assigning each stone either a positive
        # or negative sign, and we want the absolute final sum to be as small as possible.
        #
        # Another way to view it:
        # split stones into two groups. If one group's sum is x, the other group's
        # sum is total - x. The final remaining weight is:
        #
        #     abs((total - x) - x) = abs(total - 2 * x)
        #
        # To minimize this difference, we want x to be as close to total / 2 as possible.
        #
        # Therefore, this becomes a 0/1 knapsack problem:
        # find the maximum subset sum that is <= total // 2.
        # We focus on the smaller group and try to maximize its sum without exceeding half.

        total = sum(stones)
        s = total // 2

        dp = {}

        def f(i, c):
            # f(i, c) means:
            # using stones[0...i], given that the current chosen sum is c,
            # return the maximum sum we can finally reach without exceeding s.
            #
            # Different from the common 416 recursion:
            # - in 416, c usually means remaining capacity / target still needed
            # - here, c means current chosen sum, starting from 0, or you can say how much capacity left

            if (i, c) in dp:
                return dp[(i, c)]

            # No stones left, so the final chosen sum is c
            if i < 0:
                return c

            # Option 1: do not put stones[i] into the smaller group
            not_take = f(i - 1, c)

            # Option 2: put stones[i] into the smaller group,
            # only if the chosen sum does not exceed half
            if c + stones[i] <= s:
                take = f(i - 1, c + stones[i])
                dp[(i, c)] = max(take, not_take)
            else:
                dp[(i, c)] = not_take

            return dp[(i, c)]

        best = f(len(stones) - 1, 0)

        # best is the largest subset sum <= total // 2.
        # The other group has sum total - best.
        # Final difference = (total - best) - best = total - 2 * best.
        return total - 2 * best

    # ==========================================

        total = sum(stones)
        s = total // 2

        dp = [[0] * (s+1) for _ in range(len(stones)+1)]

        for c in range(s + 1):
            dp[0][c] = c

        for i in range(1, len(stones)+1):
            for c in range(s+1):
                if c + stones[i-1] <= s:
                    dp[i][c] = max(dp[i-1][c], dp[i-1][c+stones[i-1]])
                else:
                    dp[i][c] = dp[i-1][c]

        best = dp[len(stones)][0]
        return total - 2 * best

    # ===========================================
        # Because, our c means current capacity, so from 2d into 1d, we do not need to reverse the loop
        total = sum(stones)
        s = total // 2

        dp = [0] * (s+1)

        for c in range(s + 1):
            dp[c] = c

        for i in range(1, len(stones)+1):
            for c in range(s+1):
                if c + stones[i-1] <= s:
                    dp[c] = max(dp[c], dp[c+stones[i-1]])

        best = dp[0]
        return total - 2 * best

    # ---------------- @2 ---------------------------
    # Last one we define c as current capacity, now we define it as total capacity

        total = sum(stones)
        s = total // 2

        dp = {}

        def f(i, c):

            if i < 0:
                return 0

            if (i, c) in dp:
                return dp[(i, c)]

            if c >= stones[i]:
                no_take = f(i-1, c)
                take = f(i-1, c - stones[i]) + stones[i] # we add stones[i], so capacity reduces by stones[i]

                best = max(no_take, take)
            else:
                best = f(i-1, c)

            dp[(i, c)] = best
            return best

        return total - 2 * f(len(stones)-1, s)

    # ============================================

        total = sum(stones)
        s = total // 2

        dp = [[0] * (s+1) for _ in range(len(stones)+1)]

        for i in range(1, len(stones)+1):
            for c in range(1, s+1):
                if c >= stones[i-1]:
                    dp[i][c] = max(dp[i-1][c], dp[i-1][c-stones[i-1]] + stones[i-1])

                else:
                    dp[i][c] = dp[i-1][c]

        return total - 2 * dp[-1][-1]

    # ============================================
        total = sum(stones)
        s = total // 2

        dp = [0] * (s+1)

        for i in range(1, len(stones)+1):
            for c in range(s, -1, -1):
                if c >= stones[i-1]:
                    dp[c] = max(dp[c], dp[c-stones[i-1]] + stones[i-1])

                else:
                    dp[c] = dp[c]

        return total - 2 * dp[-1]
````

## LC 5 | Longest Palindromic Substring

- Topic: Dynamic Programming
- Main method: `longestPalindrome`
- Link: [Longest Palindromic Substring](https://leetcode.com/problems/longest-palindromic-substring/)


````python
class Solution:
    def longestPalindrome(self, s: str) -> str:
        # Dynamic programming solution, time complexity O(n^2), space complexity O(n^2)

        # dp[i][j] means s[i:j+1] is palindromic or not
        start = 0
        end = 0
        max_len = 0

        dp = [[False] * len(s) for _ in range(len(s))]

        for i in range(len(s)):
            dp[i][i] = True # "a" or "b" those single letter is also palindromic

        # dp[i][j] = dp[i+1][j-1] and s[i] == s[j]

        for i in range(len(s)-2, -1, -1):
            for j in range(i+1, len(s)):
                if s[i] == s[j]:
                    if j - i <= 2: # When s[i:j+1] is "a", "aa", "aba", then directly True
                        dp[i][j] = True
                    else:
                        dp[i][j] = dp[i+1][j-1]

                else:
                    dp[i][j] = False

                if dp[i][j] and j - i + 1 > max_len:
                    start = i
                    end = j
                    max_len = end-start + 1

        return s[start:end+1]

    # ===========================================
        ans_left = 0
        ans_right = 0

        # Odd situation
        for i in range(len(s)):
            l = i
            r = i

            while l >= 0 and r < len(s) and s[l] == s[r]:
                # When this stop, l and r is beyond the palindromic index, so actual index for palindromic is l+1 and r-1
                l-=1
                r+=1

            # No more palindromic, end and compare length of subtring
            if (r-1)-(l+1)+1 > ans_right-ans_left+1:
                ans_right = r-1
                ans_left = l+1

        # Even situation
        for i in range(len(s)):
            l = i
            r = i+1

            while l >= 0 and r < len(s) and s[l] == s[r]:
                # When this stop, l and r is beyond the palindromic index, so actual index for palindromic is l+1 and r-1
                l-=1
                r+=1

            # No more palindromic, end and compare length of subtring
            if (r-1)-(l+1)+1 > ans_right-ans_left+1:
                ans_right = r-1
                ans_left = l+1

        return s[ans_left:ans_right+1]
````

## LC 300 | Longest Increasing Subsequence

- Topic: Dynamic Programming
- Main method: `lengthOfLIS`
- Link: [Longest Increasing Subsequence](https://leetcode.com/problems/longest-increasing-subsequence/)


````python
class Solution:
    def lengthOfLIS(self, nums: List[int]) -> int:
        # In the greedy method, the while loop take O(n) time, so total time complexity is O(n^2)
        # We can use binary search to optimize the while loop to O(log n), so total time complexity is O(n log n)
        sub = [nums[0]]

        for num in nums[1:]:
            if num > sub[-1]:
                sub.append(num)
            else:
                # Find the first element in sub that is greater than or equal to num
                l = 0
                r = len(sub) - 1
                while l <= r:
                    mid = (l + r) // 2
                    if sub[mid] < num:
                        l = mid + 1
                    else:
                        r = mid - 1
                sub[l] = num

        return len(sub)


        # Greedy, O(n^2)
        sub = [nums[0]]

        for num in nums[1:]:
            if num > sub[-1]:
                sub.append(num)
            else:
                # Find the first element in sub that is greater than or equal to num
                i = 0
                while sub[i] < num:
                    i += 1
                sub[i] = num

        return len(sub)


        #======================================

        # Very initiative so no need recursion version
        # Time complexity is O(n^2)
        dp = [1] * len(nums)

        for i in range(len(nums)):
            for j in range(i):
                if nums[j] < nums[i]:
                    dp[i] = max(dp[i], dp[j] + 1)

        return max(dp)
````

````python
class Solution:
    def minimumTotal(self, triangle: List[List[int]]) -> int:
        dp = triangle[-1]

        for i in range(len(triangle)-2, -1, -1):
            for j in range(len(triangle[i])):
                dp[j] = min(dp[j], dp[j+1]) + triangle[i][j]

        return dp[0]

        # Bottom up, O(n^2), O(n^2) space

        dp = triangle.copy()

        for i in range(len(triangle)-2, -1, -1):
            for j in range(len(triangle[i])):
                dp[i][j] += min(dp[i+1][j], dp[i+1][j+1])

        return dp[0][0]

        #===========================

        # Top down, O(n^2), O(n^2) space

        dp = triangle.copy()

        for i in range(1, len(triangle)):
            for j in range(len(triangle[i])):
                if j == 0:
                    dp[i][j] += triangle[i-1][0]
                elif j < len(triangle[i-1]):
                    dp[i][j] += min(triangle[i-1][j], triangle[i-1][j-1])
                else:
                    dp[i][j] += triangle[i-1][j-1]

        return min(dp[-1])
````

````python
class Solution:
    def combinationSum4(self, nums: List[int], target: int) -> int:
        # Compare to 518
        f = [1] + [0] * target

        for i in range(1, target+1):
            for x in nums:
                if i - x >= 0:
                    f[i] += f[i-x]

        return f[-1]


        # Why we cannot use idea like select or not select here?
        # Because [1, 2, 3] and [3, 2, 1] are same solution here, but select or not select will count them as different solution

        dp = {}

        def f(amount):
            if amount in dp:
                return dp[amount]
            # Why amount == 0 return 1? Just like if target == 0, no selection is also one combination
            if amount == 0:
                return 1

            ans = 0
            for i in nums:
                if amount - i >= 0:
                    ans += f(amount - i)

            dp[amount] = ans
            return ans

        return f(target)
````

````python
class Solution:
    def twoSum(self, numbers: List[int], target: int) -> List[int]:
        left = 0
        right = len(numbers) - 1

        while left <= right:
            s = numbers[left] + numbers[right]
            if s > target:
                right -= 1

            elif s < target:
                left += 1

            else:
                return [left+1, right+1]
````

- Link: [3Sum](https://leetcode.com/problems/3sum/)


````python
class Solution:
    def threeSum(self, nums: List[int]) -> List[List[int]]:
        left = 0
        ans = []
        nums.sort()

        while left < len(nums)-1:
            if left > 0 and nums[left] == nums[left - 1]:
                left += 1
                continue

            mid = left + 1
            right = len(nums) - 1
            while mid < right:
                s = nums[left] + nums[mid] + nums[right]
                if s == 0:
                    ans.append([nums[left], nums[mid], nums[right]])
                    while mid < right and nums[mid] == nums[mid + 1]:
                        mid += 1

                    while mid < right and nums[right] == nums[right - 1]:
                        right -= 1

                    mid += 1
                    right -= 1

                elif s > 0:
                    right -= 1
                else:
                    mid += 1
            left += 1
        return ans
````

````python
class Solution:
    def isSubsequence(self, s: str, t: str) -> bool:
        if not s:
            return True

        p = 0

        for i in range(len(t)):
            if s[p] == t[i]:
                p += 1

            if p == len(s):
                return True

        return False

        # dp
        dp = [[0] * (len(t)+1) for _ in range(len(s)+1)]

        for i in range(1, len(s)+1):
            for j in range(1, len(t)+1):
                if s[i-1] == t[j-1]:
                    dp[i][j] = dp[i-1][j-1] + 1

                else:
                    dp[i][j] = max(dp[i - 1][j], dp[i][j - 1])

        return dp[-1][-1] == len(s)

        # =======================

        dp = [0] * (len(t)+1)

        for i in range(1, len(s)+1):
            pre = 0
            for j in range(1, len(t)+1):
                temp = dp[j]

                if s[i-1] == t[j-1]:
                    dp[j] = pre + 1

                else:
                    dp[j] = max(dp[j], dp[j - 1])
                pre = temp

        return dp[-1] == len(s)

    # =================================

        # dp[j] initially represents dp[0][j]
        # empty s is a subsequence of any t[:j]
        dp = [True] * (len(t) + 1)

        for i in range(1, len(s) + 1):
            prev = dp[0]   # old dp[i-1][0]
            dp[0] = False  # dp[i][0] = False for non-empty s[:i]

            for j in range(1, len(t) + 1):
                tmp = dp[j]  # old dp[i-1][j]

                if s[i - 1] == t[j - 1]:
                    dp[j] = prev
                else:
                    dp[j] = dp[j - 1]

                prev = tmp

        return dp[-1]

          # =========================

        dp = [[False] * (len(t)+1) for _ in range(len(s)+1)]
        for j in range(len(t) + 1):
            dp[0][j] = True

        for i in range(1, len(s)+1):
            for j in range(1, len(t)+1):
                if s[i-1] == t[j-1]:
                    dp[i][j] = dp[i-1][j-1]
                else:
                    dp[i][j] = dp[i][j-1]

        return dp[-1][-1]
    # =========================

        @cache
        def f(i, j):
            if j < 0 and i >= 0:
                return False
            if i < 0:
                return True

            if s[i] == t[j]:
                return f(i-1, j-1)
            else:
                return f(i, j-1)

        return f(len(s)-1, len(t)-1)
````

## LC 977 | Squares of a Sorted Array

````python
class Solution:
    def sortedSquares(self, nums: List[int]) -> List[int]:
        i = 0
        n = j = len(nums)-1
        ans = [0]*(n+1)

        while i <= j:
            if abs(nums[i]) < abs(nums[j]):
                ans[n] = nums[j]*nums[j]
                j-=1
            else:
                ans[n] = nums[i]*nums[i]
                i+=1
            n-=1
        return ans
````

- Topic: Binary Search
- Main method: `searchRange`
- Link: [Find First and Last Position of Element in Sorted Array](https://leetcode.com/problems/find-first-and-last-position-of-element-in-sorted-array/)


````python
class Solution:
    def searchRange(self, nums: List[int], target: int) -> List[int]:

        def bs(nums, target):

            left = 0
            right = len(nums) - 1

            while left <= right:
                mid =(right - left) // 2 + left
                if nums[mid] >= target:
                    right = mid - 1
                else:
                    left = mid + 1
            return left

        start = bs(nums, target)
        end = bs(nums, target + 1)
        return [start, end-1] if start <= end - 1 else [-1, -1]
````

## LC 33 | Search in Rotated Sorted Array

- Topic: Binary Search
- Main method: `search`
- Link: [Search in Rotated Sorted Array](https://leetcode.com/problems/search-in-rotated-sorted-array/)


````python
class Solution:
    def search(self, nums: List[int], target: int) -> int:
        # Divide array into two parts using mid
        # One part must be sorted, the other part then is rotated (Unless the original array is not rotated)
        # If the sorted part contains target so normal binary search

        # Else target must be in the other half (which may be rotated or sorted)
        #       We continue dividing, and the rotated part will again split into
        #       one sorted and one rotated part


        left = 0
        right = len(nums) - 1

        while left <= right:
            mid = (left + right) // 2

            if nums[mid] == target:
                return mid
            if nums[left] <= nums[mid]:

                if nums[left] <= target <= nums[mid]: # This does not use slice so no extra space
                    right = mid - 1
                else:
                    left = mid + 1

            else:
                if nums[mid] <= target <= nums[right]:
                    left = mid + 1
                else:
                    right = mid - 1
        return -1

        # ==========================================

        def bs(nums, target):
            left = 0
            right = len(nums) - 1

            while left <= right:
                mid = (right - left) // 2 + left
                if nums[mid] < target:
                    left = mid + 1
                elif nums[mid] > target:
                    right = mid - 1
                else:
                    return mid
            return -1

        left = 0
        right = len(nums) - 1

        while left <= right:
            mid = (right - left) // 2 + left
            if nums[left] <= nums[mid]: # Left part is sorted
                if nums[left] <= target <= nums[mid]: # Target is in left part
                    idx = bs(nums[left:mid+1], target) # This uses slice, extra space O(n)
                    return left + idx if idx != -1 else -1

                else: # Target is in right part, but right part is not sorted
                    left = mid + 1

            else: # Right part is sorted
                if nums[mid] <= target <= nums[right]: # target is in right part
                    idx = bs(nums[mid:right+1], target)
                    return mid + idx if idx != -1 else -1
                else:
                    right = mid - 1

        return -1
````

````python
class Solution:
    def searchMatrix(self, matrix: List[List[int]], target: int) -> bool:

        def bs(nums: List[int], target: int) -> int:

            left = 0
            right = len(nums) - 1

            while left <= right:
                mid = left + (right - left) // 2
                if nums[mid] > target:
                    right = mid - 1
                elif nums[mid] < target:
                    left = mid + 1
                else:
                    return True
            return False

        left = 0
        right = len(matrix) - 1

        while left <= right:
            mid = (right-left) // 2 + left

            if matrix[mid][0] <= target and matrix[mid][-1] >= target:
                return bs(matrix[mid], target)
            elif matrix[mid][0] > target:
                right = mid - 1
            else:
                left = mid + 1
        return False
````

````python
class Solution:
    def permuteUnique(self, nums: List[int]) -> List[List[int]]:
        ans = []
        path = []
        cache = [False] * len(nums)
        nums.sort()

        def f():
            if len(nums) == len(path):
                ans.append(path[:])
                return

            for j in range(len(nums)):
                if cache[j]:
                    continue

                if j > 0 and nums[j] == nums[j-1] and not cache[j-1]:
                    continue

                cache[j] = True
                path.append(nums[j])
                f()
                path.pop()
                cache[j] = False

        f()
        return ans
````

````python
class Solution:
    def letterCasePermutation(self, s: str) -> List[str]:
        path = []
        ans = []

        def f(i):
            if len(path) == len(s):
                ans.append("".join(path))
                return

            for j in range(i, len(s)):
                if s[j].isalpha():
                    path.append(s[j].upper())
                    f(j+1)
                    path.pop()

                    path.append(s[j].lower())
                    f(j+1)
                    path.pop()
                else:
                    path.append(s[j])
                    f(j+1)
                    path.pop()

        f(0)
        return ans
````

## LC 79 | Word Search

- Main method: `exist`
- Link: [Word Search](https://leetcode.com/problems/word-search/)


````python
class Solution:
    def exist(self, board: List[List[str]], word: str) -> bool:

        # If there are 4 "a"s in board, but 5 "a"s in word, we can directly return False
        board_count = Counter(c for row in board for c in row)
        word_count = Counter(word)

        for ch in word_count:
            if word_count[ch] > board_count.get(ch, 0):
                return False

        m, n = len(board), len(board[0])

        def f(i, j, k):
            if board[i][j] != word[k]:
                return False
            if k == len(word) - 1:
                return True

            tmp = board[i][j]
            board[i][j] = float("inf")

            if (j - 1 >= 0 and f(i, j - 1, k + 1)) or \
            (j + 1 < n and f(i, j + 1, k + 1)) or \
            (i - 1 >= 0 and f(i - 1, j, k + 1)) or \
            (i + 1 < m and f(i + 1, j, k + 1)):
                return True
            else:
                # Not found
                board[i][j] = tmp
                return False

        for i in range(m):
            for j in range(n):
                # We try to start from all index
                if f(i, j, 0):
                    return True
        return False
    # =====================================
        m = len(board)
        n = len(board[0])

        def f(i, j, k):
            if board[i][j] == word[k]:
                if k == len(word) - 1:
                    return True
            else:
                return False

            tmp = board[i][j]
            board[i][j] = float("inf")

            left = right = up = down = False
            if i - 1>= 0:
                left = f(i-1, j, k+1)

            if j - 1 >= 0:
                up = f(i, j-1, k+1)

            if i + 1 < m:
                right = f(i+1, j, k+1)

            if j + 1 < n:
                down = f(i, j+1, k+1)

            if left or right or up or down:
                return True

            else:
                board[i][j] = tmp
                return False

        for i in range(len(board)):
            for j in range(len(board[0])):
                if f(i, j, 0):
                    return True

        return False

    # =====================================
````

## LC 560 | Subarray Sum Equals K

- Topic: Prefix Sum
- Main method: `subarraySum`
- Link: [Subarray Sum Equals K](https://leetcode.com/problems/subarray-sum-equals-k/)


````python
class Solution:
    def subarraySum(self, nums: List[int], k: int) -> int:
        ans = 0

        s = [0] * (len(nums)+1)

        for i in range(len(nums)):
            s[i+1] = s[i] + nums[i]

        seen = defaultdict(int)

        for j in range(len(s)):
            # if s[j] - k in seen: Default is 0, so no need to check
            ans+=seen[s[j] - k]

            seen[s[j]]+=1

        return ans
````

````python
class Solution:
    def productExceptSelf(self, nums: List[int]) -> List[int]:
        # The space complexity is O(1) (The question states that the output array does not count as extra space)
        right_or_ans = [1] * len(nums)

        for i in range(len(nums)-2,  -1, -1):
            right_or_ans[i] = right_or_ans[i+1] * nums[i+1]

        pre = 1
        for j in range(len(nums)):
            right_or_ans[j] = right_or_ans[j] * pre
            pre *= nums[j]

        return right_or_ans
    # =====================================
        # The space complexity is O(n), and we can combine the second and third loop to reduce the space complexity to O(1) as above
        n = len(nums)
        ans = []
        left = [1] * n
        right = [1] * n

        for i in range(1, len(nums)):
            left[i] = left[i-1] * nums[i-1]

        for j in range(len(nums) - 2, -1, -1):
            right[j] = right[j+1] * nums[j+1]

        for x in range(len(nums)):
            ans.append(left[x] * right[x])

        return ans
````

````python
class Solution:
    def relativeSortArray(self, arr1: List[int], arr2: List[int]) -> List[int]:

        count = Counter(arr1)
        ans = []

        for i in arr2:
            if i in count:
                ans += count[i] * [i]
                del count[i]

        for j in sorted(count):
            ans += [j] * count[j]

        return ans
````

## LC 67 | Add Binary

- Topic: Arrays and Strings
- Main method: `addBinary`
- Link: [Add Binary](https://leetcode.com/problems/add-binary/)


````python
class Solution:
    def addBinary(self, a: str, b: str) -> str:
        i = len(a) - 1
        j = len(b) - 1
        res = ""
        carry = 0

        while i >= 0 or j >= 0:
            cur_a = int(a[i]) if i >=0 else 0
            cur_b = int(b[j]) if j >=0 else 0

            total = cur_a + cur_b + carry
            res = str(total % 2) + res

            carry = total // 2
            i-=1
            j-=1
        res = '1' + res if carry else res
        return res
````

- Link: [Valid Word](https://leetcode.com/problems/valid-word/)


````python
class Solution:
    def isValid(self, word: str) -> bool:
        if len(word) < 3:
            return False

        vowel = set("aeiouAEIOU")
        vowel_flag = False
        consonant_flag = False

        for i in word:
            if not i.isalnum():
                return False

            if i in vowel:
                vowel_flag = True

            if i not in vowel and not i.isdigit():
                consonant_flag = True

        return vowel_flag and consonant_flag
````

````python
class Solution:
    def removeDuplicates(self, nums: List[int]) -> int:
        slow = 2
        fast = 2

        while fast < len(nums):
            if nums[fast] == nums[slow-2]:
                fast += 1
            else:
                nums[slow] = nums[fast]
                slow += 1
                fast += 1

        return slow
````

````python
class Solution:
    def rotate(self, nums: List[int], k: int) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        k %= len(nums) # Avoid case like [-1], k = 2

        # Do not use [::-1], because it will create a new list
        # Do not use nums[k:] + nums[:k], because it will create a new list
        # Do not use reversed(), because it will create a new iterator
        def reverse(left, right):
            while left < right:
                nums[left], nums[right] = nums[right], nums[left]
                left += 1
                right -= 1

        reverse(0, len(nums) - 1)
        reverse(0, k-1)
        reverse(k, len(nums)-1)
````

## LC 206 | Reverse Linked List

- Topic: Linked List
- Main method: `reverseList`
- Link: [Reverse Linked List](https://leetcode.com/problems/reverse-linked-list/)


````python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def reverseList(self, head: Optional[ListNode]) -> Optional[ListNode]:
        if not head:
            return None

        cur = head
        pre = None

        while cur:
            nxt = cur.next
            cur.next = pre
            pre = cur
            cur = nxt
        return pre
````

## LC 92 | Reverse Linked List II

- Topic: Linked List
- Main method: `reverseBetween`
- Link: [Reverse Linked List II](https://leetcode.com/problems/reverse-linked-list-ii/)


````python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def reverseBetween(self, head: Optional[ListNode], left: int, right: int) -> Optional[ListNode]:

        p0 = dummy = ListNode()
        dummy.next = head

        for _ in range(left - 1):
            # For example [1, 2, 3, 4, 5], left = 2, right = 4, p0 is like p0->1->2->3->4->5
            # So after this for loop, ,p0 is point to 1, not 2, and p0.next is where we begin to reverse
            p0 = p0.next

        cur = p0.next # cur is p0.next not p0, reason is above
        pre = None

        for _ in range(right - left + 1):
            nxt = cur.next
            cur.next = pre
            pre = cur
            cur = nxt

        # For now, pre = 4->3->2
        # p0 = 1
        # nxt = 5
        p0.next.next = cur # 1 ->[] -> 5
        p0.next = pre
        return dummy.next
````

## LC 25 | Reverse Nodes in k-Group

- Topic: Linked List
- Main method: `reverseKGroup`
- Link: [Reverse Nodes in k-Group](https://leetcode.com/problems/reverse-nodes-in-k-group/)


````python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def reverseKGroup(self, head: Optional[ListNode], k: int) -> Optional[ListNode]:
        dummy = ListNode()
        dummy.next = head

        p0 = dummy

        n = 0
        cur = head
        while cur:
            n+=1
            cur = cur.next

        cur = p0.next
        pre = None
        while n >= k:
            n-=k
            for _ in range(k):
                nxt = cur.next
                cur.next = pre
                pre = cur
                cur = nxt

            nxt0 = p0.next
            p0.next.next = cur
            p0.next = pre
            p0 = nxt0

        return dummy.next
````

## LC 203 | Remove Linked List Elements

- Topic: Linked List
- Main method: `removeElements`
- Link: [Remove Linked List Elements](https://leetcode.com/problems/remove-linked-list-elements/)


````python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def removeElements(self, head: Optional[ListNode], val: int) -> Optional[ListNode]:
        cur = dummy = ListNode(next = head)

        while cur and cur.next:
            if cur.next.val == val:
                cur.next = cur.next.next
            else:
                cur = cur.next
        return dummy.next
````

## LC 141 | Linked List Cycle

- Topic: Linked List
- Main method: `hasCycle`
- Link: [Linked List Cycle](https://leetcode.com/problems/linked-list-cycle/)


````python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, x):
#         self.val = x
#         self.next = None

class Solution:
    def hasCycle(self, head: Optional[ListNode]) -> bool:
        slow = head
        fast = head

        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next

            if slow == fast:
                return True
        return False
````

````python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def addTwoNumbers(self, l1: Optional[ListNode], l2: Optional[ListNode]) -> Optional[ListNode]:
        cur = ans = ListNode()
        carry = 0

        while l1 or l2 or carry:
            v1 = l1.val if l1 else 0
            v2 = l2.val if l2 else 0
            total = v1 + v2 + carry
            carry = total // 10
            cur.next = ListNode(total % 10)
            cur = cur.next

            if l1:
                l1 = l1.next
            if l2:
                l2 = l2.next
        return ans.next
````

## LC 21 | Merge Two Sorted Lists

- Topic: Linked List
- Main method: `mergeTwoLists`
- Link: [Merge Two Sorted Lists](https://leetcode.com/problems/merge-two-sorted-lists/)


````python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def mergeTwoLists(self, list1: Optional[ListNode], list2: Optional[ListNode]) -> Optional[ListNode]:
        cur = ans = ListNode()

        while list1 and list2:
            if list1.val > list2.val:
                cur.next = list2
                list2 = list2.next

            else:
                cur.next = list1
                list1 = list1.next

            cur = cur.next
        cur.next = list1 if list1 else list2
        return ans.next
````

## LC 138 | Copy List with Random Pointer

- Topic: Linked List
- Main method: `copyRandomList`
- Link: [Copy List with Random Pointer](https://leetcode.com/problems/copy-list-with-random-pointer/)


````python
"""
# Definition for a Node.
class Node:
    def __init__(self, x: int, next: 'Node' = None, random: 'Node' = None):
        self.val = int(x)
        self.next = next
        self.random = random
"""

class Solution:
    def copyRandomList(self, head: 'Optional[Node]') -> 'Optional[Node]':
        if not head:
            return None

        dic = {}

        cur = head
        while cur:
            dic[cur] = Node(cur.val)
            cur = cur.next
        # In dic, it only contains a simple node, without next and random
        # It would be like
        #{
        #  Node_A: A->None,
        #  Node_B: B->None,
        #}
        cur = head
        while cur:
            dic[cur].next = dic.get(cur.next)
            # Why not dic[cur].next = dic[cur.next]?
            # Because if dic[cur.next] is None, it would be an error
            dic[cur].random = dic.get(cur.random)
            cur = cur.next

        return dic[head]
````

## LC 100 | Same Tree

- Topic: Tree
- Main method: `isSameTree`
- Link: [Same Tree](https://leetcode.com/problems/same-tree/)


````python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def isSameTree(self, p: Optional[TreeNode], q: Optional[TreeNode]) -> bool:
        def f(p, q):
            if not p or not q:
                return p == q

            if p.val == q.val:
                return f(p.left, q.left) and f(p.right, q.right)
            else:
                return False
        return f(p, q)


        def f(p, q):
            if not p and q:
                return False

            if p and not q:
                return False

            if not p and not q:
                return True

            if p.val == q.val:
                return f(p.left, q.left) and f(p.right, q.right)
            else:
                return False
        return f(p, q)
````

````python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def isSymmetric(self, root: Optional[TreeNode]) -> bool:
        def f(l, r):
            if not l and not r:
                return True

            if not l and r:
                return False

            if not r and l:
                return False

            if l.val != r.val:
                return False

            else:
                return f(l.left, r.right) and f(l.right, r.left)

        return f(root.left, root.right)



        def f(left, right):
            if not left or not right:
                return left == right

            if left.val == right.val:
                return f(left.left, right.right) and f(left.right, right.left)
            else:
                return False

        return f(root.left,root.right)
````

## LC 110 | Balanced Binary Tree

- Topic: Tree
- Main method: `isBalanced`
- Link: [Balanced Binary Tree](https://leetcode.com/problems/balanced-binary-tree/)


````python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def isBalanced(self, root: Optional[TreeNode]) -> bool:
        def f(root):
            if not root:
                return 0

            left = f(root.left)
            right = f(root.right)

            if left == -1 or right == -1 or abs(right - left) > 1:
                return -1

            return max(left, right)+1

        return f(root) != -1
````

## LC 199 | Binary Tree Right Side View

- Topic: Tree
- Main method: `rightSideView`
- Link: [Binary Tree Right Side View](https://leetcode.com/problems/binary-tree-right-side-view/)


````python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def rightSideView(self, root: Optional[TreeNode]) -> List[int]:
        ans = []
        depth = 0

        def f(root, depth):
            if not root:
                return

            if len(ans) == depth:
                ans.append(root.val)

            f(root.right, depth+1)
            f(root.left, depth+1)

        f(root, depth)
        return ans
````

## LC 104 | Maximum Depth of Binary Tree

- Topic: Tree
- Main method: `maxDepth`
- Link: [Maximum Depth of Binary Tree](https://leetcode.com/problems/maximum-depth-of-binary-tree/)


````python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def maxDepth(self, root: Optional[TreeNode]) -> int:

        def f(node):
            if not node:
                return 0

            l = f(node.left)
            r = f(node.right)

            return max(l, r) + 1

        return f(root)
````

## LC 20 | Valid Parentheses

- Topic: Stack
- Main method: `isValid`
- Link: [Valid Parentheses](https://leetcode.com/problems/valid-parentheses/)


````python
class Solution:
    def isValid(self, s: str) -> bool:
        mapping = {
            '(': ')',
            '[': ']',
            '{': '}'
        }

        stack = []

        for i in s:
            if i in mapping:
                stack.append(i)
                continue

            if stack and mapping[stack[-1]] == i:
                stack.pop()
            else:
                return False

        return not stack
````

## LC 1249 | Minimum Remove to Make Valid Parentheses

- Topic: Stack
- Main method: `minRemoveToMakeValid`
- Link: [Minimum Remove to Make Valid Parentheses](https://leetcode.com/problems/minimum-remove-to-make-valid-parentheses/)


````python
class Solution:
    def minRemoveToMakeValid(self, s: str) -> str:
        stack = []
        remove_set = set()

        for i, k in enumerate(s):
            if k == '(':
                stack.append(i)

            if k == ')':
                if stack:
                    stack.pop()
                else:
                    remove_set.add(i)

        if not stack and not remove_set:
            return s

        if stack:
            remove_set.update(stack)

        result = []

        for i, ch in enumerate(s):
            if i not in remove_set:
                result.append(ch)

        return "".join(result)
````

## LC 166 | Fraction to Recurring Decimal

- Topic: Hash Map
- Main method: `fractionToDecimal`
- Link: [Fraction to Recurring Decimal](https://leetcode.com/problems/fraction-to-recurring-decimal/)


````python
class Solution:
    def fractionToDecimal(self, numerator: int, denominator: int) -> str:
        if not numerator:
            return "0"

        ans = []

        if (numerator < 0) != (denominator < 0):
            ans.append("-")

        numerator = abs(numerator)
        denominator = abs(denominator)
        ans.append(str(numerator//denominator))

        if numerator % denominator == 0:
            return "".join(ans)

        else:
            remainer = numerator % denominator

        ans.append('.')

        repeat = {}

        while remainer:
            if remainer in repeat:
                ans.insert(repeat[remainer], "(")
                ans.append(")")
                remainer = 0
            else:
                repeat[remainer] = len(ans)
                remainer = remainer*10
                ans.append(str(remainer//denominator))
                remainer = remainer % denominator

        return "".join(ans)
````

## LC 36 | Valid Sudoku

- Topic: Hash Set
- Main method: `isValidSudoku`
- Link: [Valid Sudoku](https://leetcode.com/problems/valid-sudoku/)


````python
class Solution:
    def isValidSudoku(self, board: List[List[str]]) -> bool:
        rows = [set() for _ in range(9)]
        # Do not use [set() * 9] that means the element in set multiply by 9
        cols = [set() for _ in range(9)]
        subbox = [[set() for _ in range(3)]for _ in range(3)]

        for r in range(len(board)):
            for c in range(len(board[0])):
                val = board[r][c]
                if val == '.':
                    continue

                # Check rows
                if val in rows[r]:
                    return False
                rows[r].add(val)

                # Check cols
                if val in cols[c]:
                    return False
                cols[c].add(val)

                # Check box
                if val in subbox[r//3][c//3]:
                    return False
                subbox[r//3][c//3].add(val)

        return True
````

- Link: [Merge Intervals](https://leetcode.com/problems/merge-intervals/)


````python
class Solution:
    def merge(self, intervals: List[List[int]]) -> List[List[int]]:
        intervals.sort(key = lambda x: x[0])

        ans = []

        for interval in intervals:
            if not ans or interval[0] > ans[-1][1]:
                ans.append(interval)

            else:
                ans[-1][1] = max(ans[-1][1], interval[1])

        return ans
````

## LC 57 | Insert Interval

- Topic: Intervals
- Main method: `insert`
- Link: [Insert Interval](https://leetcode.com/problems/insert-interval/)


````python
class Solution:
    def insert(self, intervals: List[List[int]], newInterval: List[int]) -> List[List[int]]:

        ans = []
        n = len(intervals)
        i = 0

        while i < n and intervals[i][1] < newInterval[0]:
            ans.append(intervals[i])
            i+= 1

        while i < n and intervals[i][0] <= newInterval[1]:
            newInterval[0] = min(newInterval[0], intervals[i][0])
            newInterval[1] = max(newInterval[1], intervals[i][1])
            i+=1

        ans.append(newInterval)

        while i < n:
            ans.append(intervals[i])
            i+=1

        return ans
````

## LC 452 | Minimum Number of Arrows to Burst Balloons

- Topic: Intervals
- Main method: `findMinArrowShots`
- Link: [Minimum Number of Arrows to Burst Balloons](https://leetcode.com/problems/minimum-number-of-arrows-to-burst-balloons/)


````python
class Solution:
    def findMinArrowShots(self, points: List[List[int]]) -> int:
        # This idea is, sort by end point, then for the first point, we shoot on end which can COVER THE MOST POINTS, then we move to the next point which is not covered by this shot, and shoot on its end, and so on.
        ans = 0
        pre = -float("inf")
        points.sort(key = lambda x: x[1])

        for start, end in points:
            if start > pre:
                ans += 1
                pre = end

        return ans
````

## LC 3477 | Fruits Into Baskets II

- Topic: Greedy
- Main method: `numOfUnplacedFruits`
- Link: [Fruits Into Baskets II](https://leetcode.com/problems/fruits-into-baskets-ii/)


````python
class Solution:
    def numOfUnplacedFruits(self, fruits: List[int], baskets: List[int]) -> int:
        ans = 0

        for fruit in fruits:
            left = 0
            right = left
            while right < len(baskets):
                if fruit <= baskets[right]:
                    baskets[right] = 0
                    left += 1
                    right = left
                    ans += 1
                    break
                else:
                    right += 1

        return len(fruits) - ans
````

## LC 122 | Best Time to Buy and Sell Stock II

- Topic: Greedy
- Main method: `maxProfit`
- Link: [Best Time to Buy and Sell Stock II](https://leetcode.com/problems/best-time-to-buy-and-sell-stock-ii/)


````python
class Solution:
    def maxProfit(self, prices: List[int]) -> int:
        # Calculate the difference between each day, and add up all positive differences
        # Take ex :- [1,4,7,8,6,4]
        # if you take (1, 8) , diff = 7
        # or if you take (1, 4), (4, 7), (7, 8), diff = 3 + 3 + 1 = 7

        # Going directly to 8, or going to 8 by adding all differences in between is same in result, so rather than thinking to jump, think it in this way.
        diff = []

        for i in range(1, len(prices)):
            diff += [prices[i] - prices[i-1]]

        ans = sum(x for x in diff if x > 0)
        return ans if ans else 0
````

## LC 55 | Jump Game

- Topic: Greedy
- Main method: `canJump`
- Link: [Jump Game](https://leetcode.com/problems/jump-game/)


````python
class Solution:
    def canJump(self, nums: List[int]) -> bool:
        mx = 0 # This is the max reachable index

        for i in range(len(nums)):
            if mx < i: # If the current index is beyond the max reachable index, we cannot reach it so return False, consider [3,2,1,0,4]
                return False
            mx = max(mx, i + nums[i])
        return True
````

````python
class Solution:
    def jump(self, nums: List[int]) -> int:
        # We are gonna maintain two nxt, the current max reachable index, and the next max reachable index (Global Max)
        mx_nxt = 0 # This is the global max reachable (This global means from index 0 to i, what is the max reachable index, not 0 to end)
        cur_nxt = 0 # This is the current max reachable (This current means from the last jump, what is the max reachable index)
        ans = 0

        for i in range(len(nums)-1):
            mx_nxt = max(mx_nxt, i+nums[i])
            if i == cur_nxt: # If i equals to where we already reach, that means we reach the boundary, and we cannot move forward anymore, so we need to jump, but jump from where and to where?
                cur_nxt = mx_nxt # This line means: I cannot move forward anymore, so I GO BACK to where I can jump to the global max reachable, and I jump to it, so ans += 1
                ans += 1
        return ans

    # Here is some possible confusion:
    # 1. Why for i in range(len(nums)-1): not for i in range(len(nums)):
    # Because when we are at the last index, we do not need to jump anymore, so we do not need to consider the last index.

    # 2. Why greedy can find the optimal solution?
    # Because we are not greedy about single index, but all indices from last jump to current index, we are greedy about the global max reachable from all these indices, so we can always find the optimal solution.
````

## LC 134 | Gas Station

- Topic: Greedy
- Main method: `canCompleteCircuit`
- Link: [Gas Station](https://leetcode.com/problems/gas-station/)


````python
class Solution:
    def canCompleteCircuit(self, gas: List[int], cost: List[int]) -> int:
        # There are several things needed to be considered:
        # 1. If the total gas is less than total cost, we cannot complete the circuit
        # 2. When we reach the bottom, we cannot go down more but go up, so once we find the bottom index, we start from the next index, and we cannot go lower.
        # 3. If total gas >= total cost, there must be a solution, and it is unique.


        # =====================================

        # This is O(n^2) which is TLE
        n = len(gas)
        left = 0
        right = left + n - 1

        gas = gas + gas
        cost = cost + cost


        while left < n:
            use = 0
            ok = True
            for i in range(left, right + 1):
                use += gas[i] - cost[i]
                if use < 0:
                    ok = False
                    break
            if ok:
                return left
            left += 1
            right += 1

        return -1
````

## LC 455 | Assign Cookies

- Topic: Greedy
- Main method: `findContentChildren`
- Link: [Assign Cookies](https://leetcode.com/problems/assign-cookies/)


````python
class Solution:
    def findContentChildren(self, g: List[int], s: List[int]) -> int:
        # We are going to give the min cookie to the min child, so that we can save more cookies for other children, so we are greedy about the min child and the min cookie
        g.sort()
        s.sort()

        i = 0

        for j in range(len(s)):
            if i < len(g) and s[j] >= g[i]:
                i += 1

        return i
````

## LC 48 | Rotate Image

- Topic: Matrix
- Main method: `rotate`
- Link: [Rotate Image](https://leetcode.com/problems/rotate-image/)


````python
class Solution:
    def rotate(self, matrix: List[List[int]]) -> None:
        """
        Do not return anything, modify matrix in-place instead.
        """
        # For index [i, j], after rotation it would be [j, n-1-i]
        # To achieve this, we can do:
        # 1. Transpose [i, j] -> [j, i]
        # 2. Reverse rows [j, i] -> [j, n-1-i]

        n = len(matrix)

        for i in range(n):
            for j in range(i):
                matrix[i][j], matrix[j][i] = matrix[j][i], matrix[i][j]

        for i in matrix:
            i = i.reverse()
````

````python
class Solution:
    def isPalindrome(self, x: int) -> bool:
        if x < 0 or x == 10:
            return False

        if x < 10:
            return True

        reverse = 0
        original = x
        while x != 0:
            reverse = reverse * 10 + x % 10
            x = x // 10
        return original == reverse
````

## LC 146 | LRU Cache

- Topic: Design
- Main method: `__init__`
- Link: [LRU Cache](https://leetcode.com/problems/lru-cache/)


````python
class DListNode:
    def __init__(self, key, val):
        self.key = key
        self.val  = val
        self.next = None
        self.prev = None


class LRUCache:
    def __init__(self, capacity: int):
        self.capacity = capacity
        self.dic = {}
        self.head = DListNode(-1, -1)
        self.tail = DListNode(-1, -1)
        self.head.next = self.tail
        self.tail.prev = self.head


    def get(self, key: int) -> int:
        if key not in self.dic:
            return -1

        node = self.dic[key]
        self.remove(node)
        self.add(node)
        return node.val

    def put(self, key: int, value: int) -> None:
        if key in self.dic:
            old_node = self.dic[key]
            self.remove(old_node)

        node = DListNode(key, value)
        self.dic[key] = node
        self.add(node)

        if len(self.dic) > self.capacity:
            node_to_delete = self.head.next
            self.remove(node_to_delete)
            del self.dic[node_to_delete.key]

    def add(self, node):
        previous_end = self.tail.prev
        previous_end.next = node
        node.prev = previous_end
        node.next = self.tail
        self.tail.prev = node

    def remove(self, node):
        node.prev.next = node.next
        node.next.prev = node.prev



# Your LRUCache object will be instantiated and called as such:
# obj = LRUCache(capacity)
# param_1 = obj.get(key)
# obj.put(key,value)
````

````python
# """
# This is the interface that allows for creating nested lists.
# You should not implement it, or speculate about its implementation
# """
#class NestedInteger:
#    def __init__(self, value=None):
#        """
#        If value is not specified, initializes an empty list.
#        Otherwise initializes a single integer equal to value.
#        """
#
#    def isInteger(self):
#        """
#        @return True if this NestedInteger holds a single integer, rather than a nested list.
#        :rtype bool
#        """
#
#    def add(self, elem):
#        """
#        Set this NestedInteger to hold a nested list and adds a nested integer elem to it.
#        :rtype void
#        """
#
#    def setInteger(self, value):
#        """
#        Set this NestedInteger to hold a single integer equal to value.
#        :rtype void
#        """
#
#    def getInteger(self):
#        """
#        @return the single integer that this NestedInteger holds, if it holds a single integer
#        The result is undefined if this NestedInteger holds a nested list
#        :rtype int
#        """
#
#    def getList(self):
#        """
#        @return the nested list that this NestedInteger holds, if it holds a nested list
#        The result is undefined if this NestedInteger holds a single integer
#        :rtype List[NestedInteger]
#        """

class Solution:
    def depthSum(self, nestedList: List[NestedInteger]) -> int:
        # DFS
        def dfs(ni, depth):
            ans = 0
            for i in ni:
                if i.isInteger():
                    ans += i * depth
                else:
                    ans += dfs(i.getList(), depth+1)
            return ans

        return dfs(nestedList, 1)
        # BFS
        q = deque(nestedList)
        depth = 1
        ans = 0

        while q:
            for _ in range(len(q)):
                curr = q.pop()
                if curr.isInteger():
                    ans += curr.getInteger() * depth

                else:
                    q.extendleft(curr.getList())

            depth += 1
        return ans
````

````python
'''
Josephus Problem

n, k,
f(n, k) f(n-1, k), f(n-1, k)f(n, k)

n, k, , k % n,
n-1, k % n, k % n + 1, , ,  (, , )
In the new subproblem, friend indices shift by -k. For instance, friend 3 moves from index 3 to index 1 in the new circle so we need to k to get its initial index

f(n-1, k) + k % n

'''

class Solution:
    def findTheWinner(self, n: int, k: int) -> int:
        res = 0 #  f(1, k) = 0

        for i in range(2, n+1):
            res = (res+k) % i

        return res+1

#==========================================

        def f(n, k):
            if n == 1:#, winner, index0
                return 0

            return (f(n-1, k) + k) % n

        return f(n, k)+1 #index1, +1
````

### Barclays - Problem 3

Permutation testing is a modern, generic approach to testing statistical hypotheses. While traditional hypothesis testing is based on case-specific formulas and assumptions which don't always hold in practice, permutation testing provides a single, unified, non-parametric framework to perform all kinds of tests.

The idea is to randomly permute the data multiple times and calculate the quantity of interest each time. This way, we obtain the said quantity's distribution that captures variation that is produced by random chance. Then, we compare the quantity of interest we have observed in our data to the distribution of its random variation and calculate the p-value.

The goal of this task is to write a function that applies permutation testing to perform a chi-square test in order to compare the click-through rates of three website designs.

`"1"`, `"8"`, `"9"`, `"9"`, `"919"` and `"989"`.

````python
import numpy as np
import pandas as pd

class RandomClicksGenerator:
    def __init__(self, num_ones: int, num_zeros: int):
        self.rng = np.random.default_rng(1984)
        self.clicks = np.array([1] * num_ones + [0] * num_zeros)

    def shuffled_clicks(self) -> np.ndarray:
        return self.rng.permutation(self.clicks)

cont_table = pd.DataFrame([115, 98, 123], [115, 98, 123])
````

````python
import numpy as np
import pandas as pd
# from random_clicks_generator import RandomClicksGenerator

def chi2_perm_test(cont_table: pd.DataFrame):
    # Step 1
    clicks = cont_table.iloc[0].to_numpy(dtype=int) # we can remove to numpy, then we get a series with index, to_numpy return a numpy array, which could be easier later. numpy array is faster, and support vector calculation
    non_clicks = cont_table.iloc[1].to_numpy(dtype=int)

    expected_clicks = clicks.mean()
    expected_non_clicks = non_clicks.mean()

    # Step 2
    chi2_clicks = 0
    for click in clicks:
        chi2_clicks += (click - expected_clicks) ** 2 / expected_clicks # You can also use ((clicks - expected_clicks) ** 2 / expected_clicks).sum(). Remember numpy array? This is vector calculation

    chi2_non_clicks = 0
    for non_click in non_clicks:
        chi2_non_clicks += (non_click - expected_non_clicks) ** 2 / expected_non_clicks

    chi2_obs = chi2_non_clicks + chi2_clicks

    # Step 3
    total_clicks = clicks.sum()
    total_non_clicks = non_clicks.sum()

    gen = RandomClicksGenerator(total_clicks, total_non_clicks)

    # Step 4
    rand_chi2_obs_list = []
    group_size = len(clicks) + len(non_clicks)
    for _ in range(1000):
        permuted_list = gen.shuffled_clicks()
        a = permuted_list[:group_size]
        b = permuted_list[group_size:group_size*2]
        c = permuted_list[group_size*2:]

        rand_clicks = np.array([a.sum(), b.sum(), c.sum()])
        rand_non_clicks = np.array([group_size - a.sum(), group_size - b.sum(), group_size - c.sum()])

        expected_rand_clicks = rand_clicks.mean()
        expected_rand_non_clicks = rand_non_clicks.mean()

        rand_chi2_obs = ((rand_clicks - expected_rand_clicks) ** 2 / expected_rand_clicks).sum() + ((rand_non_clicks - expected_rand_non_clicks) ** 2 / expected_rand_non_clicks).sum()

        rand_chi2_obs_list.append(rand_chi2_obs)

    # Step 5
    p_value = (np.array(rand_chi2_obs_list) >= np.array(chi2_obs)).mean().round(3)

    return {
        "expected": [expected_clicks, expected_non_clicks],
        "chi2_observed": chi2_obs,
        "chi2_random_mean": np.array(rand_chi2_obs_list).mean(),
        "p_value": p_value,
        "are_clicks_different": p_value < 0.05
    }

print(chi2_perm_test(cont_table))
````

````python
class Allocator:

    def __init__(self, N: int):

        self.N = N

        # free lists for each block size
        self.bins = {}

        s = 1
        while s <= N:
            self.bins[s] = set()
            s = s * 2

        # initially the whole memory is free
        self.bins[N].add(0)
        # It looks like this
        # bins = {
            # 1: set(),
            # 2: set(),
            # 4: set(),
            # 8: {0}
        # }
        #Which means size of 1 to 8 free memory address

        # record allocated blocks
        self.allocd = {}

    def _ceil_pow2(self, x):
        # If we allocate 5 bytes, we actually allocate to 8 block size memory
        s = 1
        while s < x:
            s = s * 2

        return s

    def alloc(self, size: int) -> int:

        need = self._ceil_pow2(size)

        s = need
        while s <= self.N and len(self.bins[s]) == 0:
            s = s * 2 # Go find the available block, if s is 4 but no 4 blocks, go to find 8, 16, 32

        if s > self.N:
            return -1

        addr = self.bins[s].pop() # Take out this addr, like 8:{0}, so addr is 8

        # split larger blocks
        while s > need: # Much larger than we need
            s = s // 2
            self.bins[s].add(addr + s)

        self.allocd[addr] = need

        return addr

    def free(self, address: int):

        s = self.allocd.pop(address)
        a = address

        while True:

            # find buddy without XOR
            if (a // s) % 2 == 0:
                buddy = a + s
            else:
                buddy = a - s

            if buddy not in self.bins[s]:
                break

            # merge with buddy
            self.bins[s].remove(buddy)

            a = min(a, buddy)
            s = s * 2

            if s > self.N:
                break

        self.bins[s].add(a)
````

````python
def solution1(N, M):
    max_side_length = int((4*N + M) ** 0.5)

    def ok(k):
        big = min(N, (k*k) // 4)
        small = k*k - big*4
        if small <= M:
            return True
        return False

    for k in range(max_side_length, -1, -1):
        if ok(k):
            return k

    return 0

def solution(M, N):

    def can(k):
        big = min(N, (k // 2) * (k // 2))
        small = k*k - big*4
        return small <= M

    left = 0
    right = int((M + 4*N)**0.5) + 1
    ans = 0

    while left <= right:
        mid = (left + right) // 2

        if can(mid):
            ans = mid
            left = mid + 1
        else:
            right = mid - 1

    return ans

def test():
    tests = [
        # (N, M, expected)
        (0, 8, 2),   # example 1
        (3, 4, 4),   # example 2
        (18, 0, 8),  # example 3
        (3, 13, 5),  # example 4
        (1, 0, 2),   # one 2x2 brick
        (0, 1, 1),   # only one 1x1
        (0, 3, 1),   # three 1x1  max square 1
        (2, 0, 2),   # two 2x2 bricks  2x2 square
    ]

    for N, M, expected in tests:
        result = solution1(N, M)
        print(f"N={N}, M={M}, result={result}, expected={expected}")
        if result != expected:
            return False

    return True


print(test())
````

````python
def solution(S, T):
    S = [int(x) for x in S]
    T = [int(x) for x in T]
    n = len(S)
    ans = 0

    for i in range(n - 1):
        diff = (T[i] - S[i]) % 10
        S[i] = (S[i] + diff) % 10
        S[i + 1] = (S[i + 1] + diff) % 10
        ans += diff

    if S[-1] != T[-1]:
        return -1

    return ans
````
